## Business concepts and properties

The following are business concepts, not a proposed table list. Decide how they should be represented after normalization. Suggested MySQL types are included to keep the exercise focused on design.

### Season

- A generated identifier: `INT UNSIGNED`
- A league year: `SMALLINT UNSIGNED`
- A season label such as Spring or Autumn: `VARCHAR(30)`
- Start and end dates: `DATE`
- Status such as planned, active, or completed: `VARCHAR(20)`

### Club

- A generated identifier: `INT UNSIGNED`
- Official name: `VARCHAR(120)`
- Year established: `SMALLINT UNSIGNED`, optional
- Contact email: `VARCHAR(254)`
- One or more contact phone numbers: each `VARCHAR(25)`
- Postal address properties: line 1, line 2, city, state or province, postal code, and country code

### Team

- A generated identifier: `INT UNSIGNED`
- Display name: `VARCHAR(100)`
- Age division such as Under 14 or Adult: `VARCHAR(30)`
- The club that operates it
- An optional preferred home venue

### Person

- A generated identifier: `INT UNSIGNED`
- Given and family names: each `VARCHAR(60)`
- Date of birth: `DATE`, optional
- Email: `VARCHAR(254)`, optional
- Zero or more phone numbers: each `VARCHAR(25)`

A person may participate as a player, a coach, or both. Do not duplicate the person's basic details for each role.

### Venue and playing area

- A venue has a generated identifier, a `VARCHAR(120)` name, and postal address properties
- A venue contains one or more separately named playing areas
- A playing-area name is `VARCHAR(80)`; examples are Court 1 or North Field
- An area may have a `SMALLINT UNSIGNED` spectator capacity

### Seasonal player participation

- The person, team, and season involved
- Shirt number: `TINYINT UNSIGNED`, optional
- Joining date: `DATE`
- Participation status: `VARCHAR(20)`

### Coaching assignment

- The person, team, and season involved
- Coaching role: `VARCHAR(40)`
- Start date and optional end date: `DATE`

### Match

- A generated identifier: `BIGINT UNSIGNED`
- Season, home team, and away team
- Scheduled start: `DATETIME`
- Playing area, which may be unknown when the match is first planned
- Match status: `VARCHAR(20)`
- Optional non-negative home and away scores: `TINYINT UNSIGNED`

## Business rules

1. A club operates one or more teams; each team is operated by exactly one club.
2. A team may participate in many seasons, and a season contains many teams. Player and coach involvement is recorded for a particular season.
3. A person cannot have two player-participation records for the same team and season.
4. A person may have different coaching roles for the same team and season, but the same role must not be duplicated.
5. A venue may contain several playing areas. A playing-area name is unique only within its venue.
6. Every match belongs to one season and has exactly one home team and one away team. They must be different teams.
7. A playing area can host many matches, but it must not host two matches with the same scheduled start time.
8. A match may be created before a playing area is assigned.
9. A club or person may have multiple phone numbers. A phone number's type, such as mobile or office, may also be recorded as `VARCHAR(20)`.
10. Postal address details must not be copied as one repeating group into every relevant business record. Design a reusable representation.
11. Preserve historical seasonal participation. Moving to another team in a later season must not overwrite earlier records.
12. Use InnoDB, `utf8mb4`, primary keys, foreign keys, appropriate `UNIQUE` constraints, and useful `CHECK` constraints.

## Part D - Prepare the MySQL database

Run this only after completing Parts A-C. The setup deliberately recreates the exercise database so the DDL can be rerun from a clean state. It removes any existing database with the same name.

```python
import mysql.connector

MYSQL_HOSTNAME = "localhost"
MYSQL_PORT = 3306
MYSQL_USERNAME = "root"
MYSQL_PASSWORD = "root"
MYSQL_DATABASE = "sports_league_exercise"

server_connection = mysql.connector.connect(
    host=MYSQL_HOSTNAME,
    port=MYSQL_PORT,
    user=MYSQL_USERNAME,
    password=MYSQL_PASSWORD,
)
server_cursor = server_connection.cursor()
server_cursor.execute(f"DROP DATABASE IF EXISTS `{MYSQL_DATABASE}`")
server_cursor.execute(
    f"CREATE DATABASE `{MYSQL_DATABASE}` "
    "CHARACTER SET utf8mb4 COLLATE utf8mb4_unicode_ci"
)
server_cursor.close()
server_connection.close()

connection = mysql.connector.connect(
    host=MYSQL_HOSTNAME,
    port=MYSQL_PORT,
    user=MYSQL_USERNAME,
    password=MYSQL_PASSWORD,
    database=MYSQL_DATABASE,
)
print(f"Connected to {MYSQL_DATABASE}:", connection.is_connected())
```
