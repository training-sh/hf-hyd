PEP 20 is a short collection of principles that guide Python's design and encourage code that is simple, clear, and readable.

## A few important ideas

- **Beautiful is better than ugly:** format code so it is pleasant to read.
- **Explicit is better than implicit:** make the intention of the code clear.
- **Simple is better than complex:** choose the simplest solution that works.
- **Readability counts:** use meaningful names and clear structure.
- **Errors should never pass silently:** handle expected errors clearly.

## Little example: readability counts

Both calculations work, but the second version makes its purpose easier to understand.

````python
# Unclear
p = [100, 250, 50]
t = sum(p)

# Clear and explicit
item_prices = [100, 250, 50]
order_total = sum(item_prices)

print(f"Order total: {order_total}")
````

## Little example: simple is better than complex

A direct expression is clearer than unnecessary steps.

````python
numbers = [1, 2, 3, 4, 5]

# Simple and readable
even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
````
