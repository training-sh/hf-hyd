# D130 — PEP 8 and Python Coding Standards

Good code must do more than produce the correct output. It should also be easy for another developer—and your future self—to read, review, test, and change.

**Goal:** learn practical Python coding conventions through small examples.

## Why do we need coding standards?

A program is usually read many more times than it is written. Without shared conventions, every file may use different naming, spacing, layout, and organization.

Coding standards help a team:

- read unfamiliar code faster
- understand a name without tracing every value
- review the behavior instead of debating formatting
- reduce mistakes hidden by crowded code
- maintain a consistent codebase
- onboard new developers more easily

> Consistent code lowers the mental effort required to understand a program.

## The computer accepts both—but people do not read them equally

These two functions produce the same result. Compare how quickly you can understand them.

````python
def total1(p,q,d):
    return p*q-(p*q*d/100)


def calculate_discounted_total(unit_price, quantity, discount_percent):
    subtotal = unit_price * quantity
    discount = subtotal * discount_percent / 100
    return subtotal - discount


assert total1(500, 2, 10) == 900
assert calculate_discounted_total(500, 2, 10) == 900
````

The first function is valid Python, but names such as `p`, `q`, and `d` force the reader to guess. The second function communicates its purpose and calculation.

Style does not replace correct logic. It makes the logic easier to inspect.

## What is a PEP?

**PEP** stands for **Python Enhancement Proposal**.

A PEP is a design document for the Python community. Depending on its purpose, it may propose a language feature, describe a process, or document guidance.

Examples:

- **PEP 1** explains the purpose and process of PEPs.
- **PEP 8** provides a style guide for Python code.
- **PEP 20** contains the Zen of Python.
- **PEP 257** covers docstring conventions.

Official references: [PEP 1](https://peps.python.org/pep-0001/) and [PEP 8](https://peps.python.org/pep-0008/).

## What is PEP 8?

PEP 8 is the **Style Guide for Python Code**. It describes conventions for code layout, indentation, whitespace, imports, naming, comments, and programming practices.

PEP 8 is guidance, not Python syntax. Code can run while still being difficult to read.

A project may define its own style rules. When project rules differ, follow the agreed project standard consistently.

## 1. Use four spaces for indentation

Indentation shows which statements belong to a block. PEP 8 recommends four spaces per indentation level. Do not mix tabs and spaces.

````python
def shipping_fee(order_total):
    if order_total >= 3000:
        return 0
    return 120


assert shipping_fee(3000) == 0
assert shipping_fee(2999) == 120
````

## 2. Use meaningful naming conventions

| Item | Convention | Example |
|---|---|---|
| variable | `snake_case` | `order_total` |
| function | `snake_case` | `calculate_tax()` |
| class | `CapWords` | `ShoppingCart` |
| constant | `UPPER_CASE` | `TAX_RATE` |
| module | short `snake_case` | `order_reader.py` |

Choose names that describe meaning, not merely data type.

````python
TAX_RATE = 0.18


class InvoiceCalculator:
    def calculate_tax(self, taxable_amount):
        return taxable_amount * TAX_RATE


calculator = InvoiceCalculator()
assert calculator.calculate_tax(1000) == 180
````

Avoid names that are unclear or easily confused:

```python
# Avoid
l = [1, 2, 3]       # lowercase L resembles the number 1
x = 1500            # meaning is unknown
def calc(a, b):      # purpose and arguments are unclear
    return a * b
```

Short names are reasonable when their meaning is obvious and local, such as `i` in a tiny loop. Descriptive names are better for business values.

## 3. Add whitespace around operators

Use one space around assignment, comparison, and arithmetic operators. Do not add spaces immediately inside parentheses, brackets, or braces.

````python
# Harder to scan:
subtotal=500*2

# Preferred:
unit_price = 500
quantity = 2
subtotal = unit_price * quantity

items = ["Keyboard", "Mouse"]
order = {"id": 101, "status": "paid"}

assert subtotal == 1000
assert items[0] == "Keyboard"
assert order["status"] == "paid"
````

## 4. Write readable function calls

Do not put a space before the opening parenthesis of a function call. Put one space after each comma.

````python
def create_order(order_id, customer_name, status):
    return {
        "id": order_id,
        "customer": customer_name,
        "status": status,
    }


order = create_order(101, "Asha", "confirmed")
assert order["customer"] == "Asha"
````

## 5. Keep lines readable

PEP 8 recommends a maximum of 79 characters for code and 72 characters for comments and docstrings. A team may agree on a slightly longer limit.

Prefer implicit continuation inside parentheses, brackets, or braces instead of a backslash.

````python
def is_priority_order(customer_type, order_total, payment_received):
    return (
        customer_type == "premium"
        and order_total >= 5000
        and payment_received
    )


assert is_priority_order("premium", 5000, True) is True
````

Long collections and calls can also be split across lines. A trailing comma makes future additions and version-control changes cleaner.

````python
supported_categories = [
    "books",
    "electronics",
    "grocery",
    "stationery",
]

assert "electronics" in supported_categories
````

## 6. Organize imports

Imports normally appear at the top of a file. Put imports on separate lines and group them in this order:

1. Python standard library
2. third-party packages
3. local application modules

Separate the groups with a blank line.

```python
import json
from pathlib import Path

import pandas as pd

from ecommerce.models import Order
```

Avoid combining unrelated imports as `import json, pathlib`.

## 7. Use blank lines to show structure

- Surround top-level functions and classes with two blank lines.
- Use one blank line between methods in a class.
- Inside a function, use blank lines sparingly to separate logical stages.

In a notebook, separate cells also help organize ideas. In a `.py` file, blank-line conventions make the structure visible.

````python
def summarize_order(prices):
    subtotal = sum(prices)
    item_count = len(prices)

    return {
        "item_count": item_count,
        "subtotal": subtotal,
    }


summary = summarize_order([500, 750])
assert summary == {"item_count": 2, "subtotal": 1250}
````

## 8. Compare `None` and Boolean values clearly

Use `is None` or `is not None` because `None` is a singleton.

For Boolean conditions, prefer the value directly instead of comparing it with `True` or `False`.

````python
coupon_code = None
payment_received = True

if coupon_code is None:
    discount = 0

if payment_received:
    status = "confirmed"

assert discount == 0
assert status == "confirmed"
````

Avoid:

```python
if coupon_code == None:
    ...

if payment_received == True:
    ...
```

An explicit comparison can still be useful when you must distinguish the Boolean object `True` from other truthy values.

## 9. Use truthiness for collections

Empty strings, lists, tuples, dictionaries, and sets are false in a Boolean context. Prefer a direct truth check when you only need to know whether a collection is empty.

````python
def checkout(cart):
    if not cart:
        raise ValueError("Cart is empty")
    return "Order placed"


assert checkout(["Keyboard"]) == "Order placed"
````

## 10. Write focused comments

A useful comment explains **why** a decision exists. It should not merely repeat the code.

```python
# Weak: subtract the discount
total = subtotal - discount

# Useful: loyalty credit is applied after promotional discounts
total = subtotal - discount - loyalty_credit
```

Update comments when code changes. An incorrect comment is worse than no comment.

## 11. Write simple docstrings for public code

A docstring explains what a module, class, method, or function provides. PEP 257 contains the detailed docstring conventions.

````python
def calculate_item_amount(quantity, unit_price):
    """Return the amount for one order line."""
    return quantity * unit_price


assert calculate_item_amount(3, 250) == 750
print(calculate_item_amount.__doc__)
````

## 12. Catch specific exceptions

A broad `except:` can hide programming errors and system-exit signals. Catch the exception you know how to handle.

````python
def parse_quantity(text):
    try:
        return int(text)
    except ValueError:
        return 0


assert parse_quantity("5") == 5
assert parse_quantity("unknown") == 0
````

## 13. Keep functions focused

PEP 8 is mainly a style guide, but readable code also benefits from small units with one clear responsibility.

Separate calculation from printing, file access, or user input when possible. Focused functions are easier to name, reuse, and test.

````python
def calculate_subtotal(prices):
    return sum(prices)


def format_currency(amount):
    return f"₹{amount:,.2f}"


subtotal = calculate_subtotal([1500, 500])
assert format_currency(subtotal) == "₹2,000.00"
````

## Mini refactoring example

The following code runs, but it has unclear names, crowded operators, and several statements on one line:

```python
def f(p,q):
    t=p*q;if t>=2000:return t*.9
    return t
```

A clearer version follows.

````python
DISCOUNT_THRESHOLD = 2000
DISCOUNT_RATE = 0.10


def calculate_total(unit_price, quantity):
    subtotal = unit_price * quantity

    if subtotal >= DISCOUNT_THRESHOLD:
        return subtotal * (1 - DISCOUNT_RATE)

    return subtotal


assert calculate_total(500, 2) == 1000
assert calculate_total(1000, 2) == 1800
````

## Consistency and judgment

PEP 8 explicitly values readability and consistency. Apply the rules thoughtfully:

- follow the existing project standard
- do not change working behavior merely to satisfy style
- do not force a rule when it makes code less readable
- configure the team’s formatter and style checker consistently

The goal is not to win a formatting contest. The goal is code that people can understand and maintain.

## PEP 8 review checklist

Before sharing Python code, check:

- [ ] four spaces are used for each indentation level
- [ ] variables and functions use meaningful `snake_case` names
- [ ] classes use `CapWords` and constants use `UPPER_CASE`
- [ ] operators and commas have readable spacing
- [ ] long expressions are wrapped inside parentheses
- [ ] imports are at the top, separated, and grouped
- [ ] blank lines make the code structure visible
- [ ] `None` is compared with `is` or `is not`
- [ ] comments explain why and remain accurate
- [ ] specific exceptions are caught
- [ ] the project’s own agreed standard is followed

## Recap

- A coding standard makes a shared codebase easier to read and maintain.
- PEP means **Python Enhancement Proposal**.
- PEP 8 is Python’s main code-style guide.
- Good style includes consistent indentation, naming, whitespace, layout, imports, comments, and programming practices.
- PEP 8 supports readable code; it does not replace correct design, tests, or team judgment.
