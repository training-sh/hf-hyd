## What is a unit test?

A unit test checks one small piece of code in isolation.

```python
def calculate_total(price, quantity):
    return price * quantity
```

Here, the function is our unit.

````python
def calculate_total(price, quantity):
    return price * quantity

result = calculate_total(500, 2)
assert result == 1000
````

## Why automate tests?

A manual check must be repeated by a person.
An automated test can run after every code change.

- faster feedback
- repeatable results
- safer changes
- examples of expected behavior

## A simple testing approach

1. Read the requirement.
2. Select one behavior.
3. Choose input and expected output.
4. Run the function.
5. Compare actual and expected values.
6. Add normal, boundary, and invalid cases.

## Arrange — Act — Assert

- **Arrange:** prepare input and expected result
- **Act:** call the function
- **Assert:** compare actual and expected results

````python
price = 750              # Arrange
quantity = 2
expected = 1500

actual = calculate_total(price, quantity)  # Act
assert actual == expected                 # Assert
````

## Start with the happy path

The happy path uses normal, valid input.

````python
def apply_discount(price, percent):
    return price * (1 - percent / 100)

actual = apply_discount(1000, 10)
assert actual == 900
````

## Add boundary cases

Boundaries are values where behavior begins, ends, or changes.

For discount percentage, useful boundaries are `0` and `100`.

````python
assert apply_discount(1000, 0) == 1000
assert apply_discount(1000, 100) == 0
````

## Test each branch

Each possible result from an `if` statement needs a test.

````python
def shipping_fee(total):
    if total >= 500:
        return 0
    return 50

assert shipping_fee(500) == 0
assert shipping_fee(499) == 50
````

## Test invalid input

Invalid input should produce a defined response, usually an exception.

````python
def set_quantity(quantity):
    if quantity <= 0:
        raise ValueError("Quantity must be positive")
    return quantity

assert set_quantity(2) == 2
````

````python
try:
    set_quantity(0)
    assert False, "Expected ValueError"
except ValueError as error:
    assert str(error) == "Quantity must be positive"
````

## Common test methods

| Method | Example check |
|---|---|
| exact result | total equals `1000` |
| Boolean result | product is in stock |
| boundary | free delivery starts at `500` |
| exception | quantity `0` raises `ValueError` |
| collection | cart contains a product |
| type | returned order is a dictionary |

````python
cart = ["Keyboard", "Mouse"]
order = {"id": 101, "status": "paid"}

assert "Keyboard" in cart
assert len(cart) == 2
assert isinstance(order, dict)
````

## One behavior per test

Small tests make failures easy to understand.

```python
def test_shipping_is_free_at_500():
    assert shipping_fee(500) == 0

def test_shipping_costs_50_below_500():
    assert shipping_fee(499) == 50
```

## Test names explain behavior

Use this pattern:

```text
test_<unit>_<situation>_<expected_result>
```

Examples:

- `test_discount_with_zero_percent_keeps_price`
- `test_empty_cart_cannot_checkout`
- `test_quantity_zero_raises_value_error`

## Testing exceptions with pytest

```python
import pytest

def test_zero_quantity_is_rejected():
    with pytest.raises(ValueError):
        set_quantity(0)
```

`pytest.raises` passes only when the expected exception occurs.

## Built-in unittest

`unittest` needs no installation. Tests are methods inside a class.

````python
import unittest

class TestShop(unittest.TestCase):
    def test_total(self):
        self.assertEqual(calculate_total(500, 2), 1000)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestShop)
unittest.TextTestRunner(verbosity=1).run(suite)
````

## Useful unittest methods

| Method | Meaning |
|---|---|
| `assertEqual(a, b)` | values are equal |
| `assertTrue(value)` | value is true |
| `assertFalse(value)` | value is false |
| `assertIn(item, group)` | item is present |
| `assertIsNone(value)` | value is `None` |
| `assertRaises(Error)` | expected error occurs |

````python
case = unittest.TestCase()

case.assertTrue(8 > 0)
case.assertIn("Keyboard", cart)
case.assertIsInstance(order, dict)
with case.assertRaises(ValueError):
    set_quantity(0)
````

## Understand test results

- **Passed:** actual behavior matched expectation
- **Failed:** an assertion was false
- **Error:** test code stopped unexpectedly
- **Skipped:** test was intentionally not run

Read the first failure, fix it, and run the tests again.

## Tests should be independent

Each test prepares its own data.
It should not rely on another test running first.

````python
def test_new_cart_is_empty():
    cart = []
    assert cart == []

def test_item_can_be_added():
    cart = ["Keyboard"]
    assert len(cart) == 1

test_new_cart_is_empty()
test_item_can_be_added()
````

## What belongs in a unit test?

Prefer:

- one small function
- fixed input data
- a clear expected result
- fast execution

Avoid real payments, networks, databases, and changing external data in beginner unit tests.

## Quick practice

Test the normal result and the free-delivery boundary.

````python
def final_amount(total):
    fee = 0 if total >= 500 else 50
    return total + fee

assert final_amount(300) == 350
assert final_amount(500) == 500
````

- [ ] test normal input
- [ ] test boundaries and branches

- [ ] use descriptive test names

## Recap

- Unit tests verify small pieces of code.
- Plan inputs and expected outputs before coding.
- Python includes `unittest`.
- pytest requires installation but offers simpler syntax.
- Passing tests support confidence; they do not prove the absence of every bug.
