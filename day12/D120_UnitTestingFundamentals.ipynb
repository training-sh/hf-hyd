# D120 — Unit Testing Fundamentals

### A 30-minute introduction for Python beginners

**Goal:** learn how to check one small unit of code with clear, repeatable examples.

## Learning objectives

By the end, you can:

- explain what a unit test does
- write a test using **Arrange, Act, Assert**
- choose normal, edge, and invalid inputs
- inspect branches and conditions using white-box thinking
- name and organize small Python tests

## What is a unit?

A **unit** is a small, testable part of a program.

For beginners, think of one Python function:

```python
def add(a, b):
    return a + b
```

A unit test calls the function and checks its result.

## Why write unit tests?

- Catch mistakes early
- Explain expected behavior with examples
- Make code changes safer
- Run the same checks repeatedly

> A test is a tiny executable requirement.

## The testing mindset

For each function, ask:

1. What inputs are accepted?
2. What output is expected?
3. What decisions does the code make?
4. What happens at the limits?
5. What should happen for invalid input?

## Arrange — Act — Assert

| Step | Question |
|---|---|
| **Arrange** | What input and expected value do I need? |
| **Act** | Which function do I call? |
| **Assert** | Did I get the expected result? |

````python
def add(a, b):
    return a + b

expected = 5          # Arrange
actual = add(2, 3)    # Act
assert actual == expected  # Assert
````

## Reading an assertion

```python
assert actual == expected
```

- If the expression is `True`, the test passes.
- If it is `False`, Python raises `AssertionError`.
- A useful test checks one clear behavior.

````python
name = "Ada"
items = ["pen", "book"]

assert name.startswith("A")
assert "pen" in items
assert len(items) == 2
````

## Function arguments become test inputs

Every argument creates choices for testing.

```python
def final_price(price, discount):
    return price * (1 - discount)
```

Consider the type, allowed range, and meaning of both `price` and `discount`.

````python
def final_price(price, discount):
    return price * (1 - discount)

assert final_price(100, 0.10) == 90
assert final_price(100, 0) == 100
assert final_price(100, 1) == 0
````

## Choose input groups

Inputs that should behave alike form an **equivalence group**.

| Group | Example |
|---|---|
| normal valid value | `price=100` |
| lowest allowed value | `price=0` |
| invalid value | `price=-1` |

Test one representative from each important group.

## Edge and boundary conditions

Bugs often hide where behavior changes.

For an allowed age range of 18 through 60, test:

- just below: `17`
- lower boundary: `18`
- upper boundary: `60`
- just above: `61`

Also consider empty text, empty lists, zero, and one item.

````python
def can_apply(age):
    return 18 <= age <= 60

assert can_apply(17) is False
assert can_apply(18) is True
assert can_apply(60) is True
assert can_apply(61) is False
````

## White-box testing

**White-box testing** uses knowledge of the code inside the function.

We inspect:

- `if` / `elif` / `else` branches
- Boolean conditions
- loops and early returns
- statements that may raise errors

The goal is useful coverage, not merely a large number of tests.

## Branch coverage

A **branch** is one possible path through a decision.

```python
def shipping(total):
    if total >= 500:
        return 0
    return 50
```

We need one test where the condition is true and one where it is false.

````python
def shipping(total):
    if total >= 500:
        return 0
    return 50

assert shipping(500) == 0
assert shipping(499) == 50
````

## Multiple branches

List every possible outcome before writing tests.

```python
def label(score):
    if score >= 80:
        return "high"
    if score >= 50:
        return "medium"
    return "low"
```

There are three outcomes, so start with at least three tests.

````python
def label(score):
    if score >= 80: return "high"
    if score >= 50: return "medium"
    return "low"

assert label(80) == "high"
assert label(50) == "medium"
assert label(49) == "low"
````

## Compound conditions

```python
if has_ticket and age >= 18:
```

Important combinations are:

| `has_ticket` | `age >= 18` | Result |
|---|---|---|
| True | True | allow |
| True | False | deny |
| False | True | deny |
| False | False | deny |

````python
def can_enter(has_ticket, age):
    return has_ticket and age >= 18

assert can_enter(True, 18) is True
assert can_enter(True, 17) is False
assert can_enter(False, 18) is False
````

## Invalid inputs and exceptions

A function should have a clear rule for bad input:

- reject it with an exception, or
- return a documented special value

Do not silently guess unless that behavior is intentional.

````python
def divide(a, b):
    if b == 0:
        raise ValueError("b cannot be zero")
    return a / b

assert divide(6, 2) == 3
````

With pytest, check the expected exception:

```python
import pytest

def test_zero_divisor():
    with pytest.raises(ValueError):
        divide(6, 0)
```

The test passes only when `ValueError` is raised.

## From notebook check to test file

pytest discovers files and functions beginning with `test_`.

```python
def add(a, b):
    return a + b

def test_add_two_numbers():
    assert add(2, 3) == 5
```

Run from a terminal: `python -m pytest`

## Good test names

A name should describe the situation and expected result.

```text
test_shipping_is_free_at_500
test_age_17_cannot_apply
test_divide_rejects_zero_divisor
```

Avoid unclear names such as `test1` or `check_function`.

## Qualities of a useful unit test

A test should be:

- **small** — checks one behavior
- **clear** — intent is easy to read
- **repeatable** — same result every run
- **independent** — does not depend on another test
- **fast** — avoids real networks, databases, and long waits

## Common beginner mistakes

- Testing only the happy path
- Forgetting boundary values
- Writing many checks in one test
- Copying the function's logic into the test
- Depending on today's date, random values, or test order
- Treating coverage percentage as proof that behavior is correct

## Guided example: plan before coding

Requirement: `fee(weight)` returns ₹50 up to and including 5 kg, otherwise ₹100.

Test plan:

| Case | Input | Expected |
|---|---:|---:|
| normal lower branch | 2 | 50 |
| boundary | 5 | 50 |
| just above boundary | 6 | 100 |

````python
def fee(weight):
    if weight <= 5:
        return 50
    return 100

assert fee(2) == 50
assert fee(5) == 50
assert fee(6) == 100
````

## Quick practice

For the function below, identify tests before running it:

```python
def is_valid_quantity(qty):
    return 1 <= qty <= 10
```

1. One normal input
2. Both boundaries
3. Just outside both boundaries
4. The expected result for each input

### Suggested answer

```python
assert is_valid_quantity(5) is True
assert is_valid_quantity(1) is True
assert is_valid_quantity(10) is True
assert is_valid_quantity(0) is False
assert is_valid_quantity(11) is False
```

The exact boundaries matter more than testing many random numbers.

## Test-design checklist

Before finishing a function, check:

- [ ] normal valid inputs
- [ ] each branch or result
- [ ] boundary values and values just outside
- [ ] empty, zero, or one-item cases when relevant
- [ ] invalid types or values
- [ ] expected exceptions
- [ ] clear test names and one behavior per test

## Recap

- A unit test checks one small piece of code.
- Use **Arrange, Act, Assert**.
- Function arguments guide input selection.
- Test normal, edge, and invalid values.
- White-box testing reveals branches and conditions.
- A passing test is useful only when it checks the right behavior.
