# D121 — Assertion Basics

Assertions check whether an expected condition is true.

```python
assert condition
assert condition, "failure message"
```

## Our e-commerce example

We will test products, carts, discounts, stock, and orders.

````python
product = {
    "name": "Keyboard",
    "price": 1500,
    "stock": 8,
    "category": "electronics"
}
````

## A passing assertion

Nothing is displayed when the condition is true.

````python
actual_name = product["name"]
expected_name = "Keyboard"

assert actual_name == expected_name
````

## A failing assertion

A false condition raises **`AssertionError`**.

The failing line below is commented so **Run All** can continue.

````python
actual_price = product["price"]

# Remove # to observe AssertionError
# assert actual_price == 1200
````

## Add a useful failure message

The message should explain the expected behavior.

````python
expected_price = 1500
actual_price = product["price"]

assert actual_price == expected_price, (
    f"Expected ₹{expected_price}, got ₹{actual_price}"
)
````

## Equality and inequality

Use `==` when values must match and `!=` when they must differ.

````python
order_id = "ORD-101"
cancelled_id = "ORD-099"

assert order_id == "ORD-101"
assert order_id != cancelled_id
````

## Numeric comparisons

Use `<`, `<=`, `>`, and `>=` for ranges and limits.

````python
quantity = 2
stock = product["stock"]
price = product["price"]

assert quantity > 0
assert quantity <= stock
assert price >= 0
````

## Boolean results

For a Boolean result, explicitly check `is True` or `is False`.

````python
def is_in_stock(stock):
    return stock > 0

assert is_in_stock(8) is True
assert is_in_stock(0) is False
````

## Membership

Use `in` and `not in` for collections and text.

````python
categories = ["books", "electronics", "grocery"]
blocked_skus = {"OLD-10", "BAD-20"}

assert product["category"] in categories
assert "KEY-101" not in blocked_skus
````

## Empty and non-empty collections

Use `len()` when the exact number matters.

````python
cart = ["KEY-101", "MOU-201"]
empty_cart = []

assert len(cart) == 2
assert cart
assert not empty_cart
````

## `None` values

Use `is None` or `is not None`, not `== None`.

````python
coupon = None
order_id = "ORD-101"

assert coupon is None
assert order_id is not None
````

## Type checks

Use `isinstance()` when the output type is part of the requirement.

````python
order_total = 3000.0
cart_items = ["KEY-101"]

assert isinstance(order_total, float)
assert isinstance(cart_items, list)
````

## Strings

Check exact content or a required part of the text.

````python
status = "Order shipped"
order_code = "ORD-101"

assert status == "Order shipped"
assert "shipped" in status
assert order_code.startswith("ORD-")
````

## Lists and dictionaries

Check the complete value or only the important part.

````python
cart = ["Keyboard", "Mouse"]
order = {"id": 101, "status": "paid"}

assert cart == ["Keyboard", "Mouse"]
assert cart[0] == "Keyboard"
assert order["status"] == "paid"
````

## Floating-point values

Some decimal calculations are not stored exactly.
Compare using a small tolerance.

````python
price = 100
tax_rate = 0.18
actual_tax = price * tax_rate
expected_tax = 18.0

assert abs(actual_tax - expected_tax) < 0.001
````

## Multiple conditions

A combined assertion is valid, but separate assertions show failures more clearly.

````python
discount = 10
customer_type = "premium"

assert discount >= 0
assert discount <= 50
assert customer_type in {"regular", "premium"}
````

## Inspecting `AssertionError`

`AssertionError` is the exception raised by a failed assertion.
Normally, a test runner reports it for us.

````python
try:
    assert product["stock"] == 0, "Product should be sold out"
except AssertionError as error:
    print(type(error).__name__)
    print(error)
````

## Testing an expected exception

This basic pattern confirms that invalid input raises the correct exception.

````python
def checkout(cart):
    if not cart:
        raise ValueError("Cart is empty")
    return "Order placed"

assert checkout(["KEY-101"]) == "Order placed"
````

````python
try:
    checkout([])
    assert False, "Expected ValueError"
except ValueError as error:
    assert str(error) == "Cart is empty"
````

## Where assertions are useful

- Unit tests: verify actual and expected results
- Development checks: confirm internal assumptions
- Debugging: detect an impossible program state
- Data pipeline tests: verify transformed rows or columns

Assertions are mainly checks for developers.

````python
def cart_total(prices):
    total = sum(prices)
    assert total >= 0, "Cart total cannot be negative"
    return total

assert cart_total([1500, 500]) == 2000
````

## Do not use `assert` for user validation

Python can remove assertions when run with optimization: `python -O`.

Use normal exceptions for customer input, payment rules, permissions, and other required business checks.

````python
def set_quantity(quantity):
    if quantity <= 0:
        raise ValueError("Quantity must be positive")
    return quantity

assert set_quantity(2) == 2
````

## Assertions inside pytest tests

pytest uses Python's normal `assert` and gives a detailed failure report.

```python
def test_premium_discount():
    actual = 1000 - 100
    assert actual == 900
```

Run it with: `python -m pytest`

## Quick practice

Write assertions for this order:

```python
order = {"total": 2500, "status": "paid", "items": 3}
```

Check that the total is positive, status is paid, and item count is 3.

````python
order = {"total": 2500, "status": "paid", "items": 3}

assert order["total"] > 0
assert order["status"] == "paid"
assert order["items"] == 3
````

## Remember

- `assert` expects a true condition.
- A false condition raises `AssertionError`.
- Add a message when it improves the failure report.
- Assert values, ranges, membership, types, and expected states.
- Keep separate assertions when separate failures matter.
- Use exceptions, not assertions, for required user validation.
