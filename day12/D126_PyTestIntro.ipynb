# D126 — Introduction to pytest

`pytest` is a Python testing tool. A test is usually a small function that checks whether code produces the expected result.

## 1. Install pytest

Run this command in a terminal:

```bash
python -m pip install pytest
```

Inside a notebook, `%pip` installs into the current notebook kernel:

```python
%pip install pytest
```

Using `python -m pytest` is helpful because it runs pytest with the selected Python interpreter.

## 2. A very small test

pytest uses normal Python `assert` statements.

````python
def add(first_number, second_number):
    return first_number + second_number


def test_add():
    assert add(2, 3) == 5


# A direct call gives us a quick check inside this notebook.
test_add()
print("Smoke test passed")
````

## 3. Run this notebook as a smoke test

A **smoke test** executes the notebook from beginning to end and catches basic errors. Install the `nbmake` pytest plugin:

```bash
python -m pip install pytest nbmake
```

From the `D12_UnitTesting` folder, run only this notebook:

```bash
python -m pytest --nbmake D126_PyTestIntro.ipynb
```

Run every notebook in the folder:

```bash
python -m pytest --nbmake "*.ipynb"
```

> `nbmake` checks that notebook cells execute successfully. Put detailed unit tests in normal `.py` files so they remain easy to organize and run.

## 4. MovieLens repository structure

Assume `movielens-repo` is the repository root and `movielens` is the top-level Python package. Keep production code under `src/movielens/` and tests in a separate top-level `tests/` folder. The test folders can mirror the package folders:

```text
movielens-repo/
├── pyproject.toml
├── requirements-dev.txt
├── src/
│   └── movielens/
│       ├── __init__.py
│       ├── config/
│       │   ├── __init__.py
│       │   └── settings.py
│       ├── models/
│       │   ├── __init__.py
│       │   └── movie.py
│       ├── datasets/
│       │   ├── __init__.py
│       │   └── ratings.py
│       ├── readers/
│       │   ├── __init__.py
│       │   └── csv_reader.py
│       └── writers/
│           ├── __init__.py
│           └── csv_writer.py
└── tests/
    ├── config/
    │   └── test_settings.py
    ├── models/
    │   └── test_movie.py
    ├── datasets/
    │   └── test_ratings.py
    ├── readers/
    │   └── test_csv_reader.py
    └── writers/
        └── test_csv_writer.py
```

The `__init__.py` files make the source directories importable Python packages. Tests do not need an `__init__.py` unless the tests themselves must behave as a package. Test filenames should start with `test_`, and test function names should also start with `test_`.

### Example source: `src/movielens/models/movie.py`

```python
def format_movie_title(title, year):
    return f"{title} ({year})"
```

### Example test: `tests/models/test_movie.py`

```python
from movielens.models.movie import format_movie_title


def test_format_movie_title():
    result = format_movie_title("Toy Story", 1995)

    assert result == "Toy Story (1995)"
```

Notice that the import starts with `movielens`, not `src.movielens`. The `src` directory is a container for packages and is not part of the installed import name.

## 5. Run package tests

Open a terminal in `movielens-repo`, the repository root containing both `src` and `tests`. Install the project in editable mode once, together with pytest:

```bash
python -m pip install -e .
python -m pip install pytest
```

Then run every test:

```bash
python -m pytest
```

Useful variations:

```bash
# Show each test name
python -m pytest -v

# Run one test file
python -m pytest tests/models/test_movie.py

# Run one test function
python -m pytest tests/models/test_movie.py::test_format_movie_title
```

Always run these commands from `movielens-repo`, not from inside `src`, `movielens`, or `tests`.

## 6. Optional development requirements file

Store test-only tools in `requirements-dev.txt`:

```text
pytest
nbmake
```

Install them together with:

```bash
python -m pip install -r requirements-dev.txt
```

## Summary

1. Install pytest with `python -m pip install pytest`.
2. Put source code in `src/movielens/` and write tests in the project-level `tests/` folder.
3. Name test files and functions with the `test_` prefix.
4. Run package tests from the project root with `python -m pytest`.
5. Install `nbmake` and use `python -m pytest --nbmake notebook.ipynb` for a notebook smoke test.
