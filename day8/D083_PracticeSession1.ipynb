# Exercise — begin here

You are architecture interns for **NovaCart**, a growing e-commerce company.

Your task:

> Draft an enterprise data flow showing how NovaCart’s operational data is created, moved, prepared, and consumed.

Create the workflow in **PowerPoint**, export the completed slide as an image, and upload the image.

Do not begin by choosing product names. Begin with systems, responsibilities, data, and consumers.


# Expected submission

Submit **one architecture image** exported from PowerPoint.

The image must show:

- Data-producing applications and systems
- Operational databases
- Movement between systems
- An analytical platform
- Data preparation or transformation
- Final consumers
- Clearly labeled arrows
- A small legend

Recommended filename:

```text
TeamName_NovaCart_DataFlow.png
```

# Scenario: NovaCart

NovaCart sells electronics and household products.

Customers can shop through:

- A mobile application
- A web application

Warehouse employees use a separate application to:

- Check available stock
- Reserve stock for confirmed orders
- Record picking and packing
- Update stock after dispatch

Business teams use reports to understand sales, inventory, customers, and delivery performance.


# Business growth

NovaCart currently handles:

- 20,000 orders on a normal day
- 100,000 orders during major sales
- Three warehouses
- Thousands of simultaneous users
- Daily management reports
- Requests for fresher inventory and sales information

The company expects traffic and data volume to double next year.

Your design should work now and show where future scaling may be required.


# Actors and consumers

Include these people or systems:

### Operational actors

- Customer
- Warehouse employee
- Customer-support employee

### Analytical consumers

- Sales manager
- Inventory manager
- Business analyst
- Executive leadership

Optional: a forecasting model or customer recommendation service.


# Required operational flows

Your diagram must represent:

1. Customer browses products
2. Customer places an order
3. Payment result is recorded
4. Stock is checked and reserved
5. Warehouse picks and packs the order
6. Shipment status is updated
7. Customer support can view order status

Ask:

> Which system owns the current truth for each step?


# Required analytical questions

NovaCart wants to answer:

- What was daily and monthly revenue?
- Which products sell most by region?
- Which products frequently run out of stock?
- How long does order fulfillment take?
- Which warehouses experience delays?
- How many customers purchase again?
- What is the cancellation and refund rate?

Your analytical design must make these questions possible.


# Data you may need

| Data area | Example fields |
|---|---|
| Customer | customer ID, region, registration time |
| Product | product ID, category, price |
| Order | order ID, customer, time, status, total |
| Order item | order ID, product, quantity, price |
| Payment | order ID, result, amount, time |
| Inventory | product, warehouse, available quantity |
| Fulfillment | pick, pack, dispatch, delivery times |

You may add data, but explain why it is required.


# Architecture constraints

Your design must:

- Keep checkout responsive
- Prevent incorrect stock reservation
- Avoid running large reports on live transactional databases
- Preserve historical orders and inventory changes
- Combine mobile, web, warehouse, and payment data
- Refresh management reports at least every hour
- Protect sensitive customer and payment information
- Remain understandable to a non-technical manager


# What your diagram must contain

```text
SOURCES
  mobile app | web app | warehouse app | payment provider
       ↓
OPERATIONAL SYSTEMS
  application services and OLTP databases
       ↓
DATA MOVEMENT
  labeled batch, incremental/CDC, or streaming flows
       ↓
ANALYTICAL PLATFORM
  landing/raw → validation/transformation → warehouse/OLAP
       ↓
CONSUMERS
  reports | analysts | managers | optional models
```

Adapt this structure—do not merely copy the text.


# Design rule: boxes need responsibilities

Weak label:

```text
Database
```

Better label:

```text
Orders OLTP
Owns current orders and order status
```

Weak label:

```text
Analytics
```

Better label:

```text
Analytical warehouse
Integrated historical sales and fulfillment data
```

Every major box should make its role understandable.


# Design rule: arrows need meaning

An arrow should describe what moves and how.

Weak:

```text
Orders DB ─────────► Warehouse
```

Better:

```text
Orders OLTP ── order changes, incremental every 15 min ──► landing
```

Possible arrow labels:

- Customer request
- Transactional read/write
- Hourly incremental order changes
- Continuous inventory event
- Daily reference-data batch
- Curated analytical query


# PowerPoint setup

Create a new **blank 16:9 slide**.

Suggested layout:

```text
┌───────────┬──────────────┬──────────────┬─────────────┐
│ Sources   │ Operational  │ Movement &   │ Consumers   │
│ & users   │ systems      │ analytics    │             │
└───────────┴──────────────┴──────────────┴─────────────┘
```

Use:

- Rectangles for systems
- Cylinders for stored data
- Arrows for movement
- Colors for architectural roles
- Short labels rather than paragraphs


# Recommended visual language

| Color | Meaning |
|---|---|
| Blue | Source/application |
| Green | Operational/OLTP |
| Orange | Movement or transformation |
| Purple | Analytical/OLAP |
| Grey | Consumer |

Arrow styles:

- Solid arrow — request or data movement
- Dashed arrow — optional or future flow

Include this legend on the slide. Consistency matters more than the exact colors.

Before drawing, list:

1. Business actors

5. Analytical questions
6. Final consumers

Then draw a rough flow on paper:

```text
actor → application → operational data → analytical data → consumer
```


# Phase 1 checkpoint

Your team should be able to answer:

- Where is an order first created?
- Which system owns current inventory?
- Where is payment status recorded?
- How does the warehouse learn what to pack?
- Where is history preserved?
- Which system serves large reports?
- Who uses each report?

If ownership is unclear, the architecture is not ready to draw.


# Phase 2 — design operational systems

**Time: 20 minutes**

Show:

- Mobile and web entry points
- Order-processing responsibility
- Order OLTP data
- Inventory/warehouse responsibility
- Inventory OLTP data
- Payment interaction
- Customer-support access

Mark at least one area where **concurrent transactions** could conflict.


# Operational challenge: stock reservation

During a flash sale:

```text
Customer A ─┐
Customer B ─┼─► request the last product
Customer C ─┘
```

Your diagram should communicate:

- Which component decides whether stock is available?
- Where is the transaction recorded?
- What data must remain correct?

You do not need to show lock implementation. Add a note such as:

> “Atomic stock reservation prevents overselling.”


# Operational scaling prompt

Traffic grows ten times during sales.

Consider—not all must appear:

- Scale application services horizontally
- Keep transactional work short
- Add read replicas for suitable read traffic
- Partition large historical operational tables
- Consider sharding only if simpler scaling is insufficient

Add one **scaling note** to your diagram and state which problem it addresses.


# Phase 3 — design data movement

**Time: 20 minutes**

For every operational source moving to analytics, choose:

- Full batch
- Incremental batch
- Change Data Capture (CDC)
- Streaming events

Label:

1. What moves?
2. How frequently?
3. Where does it land?

The hourly-report requirement does not automatically require every flow to be real time.


# Movement decision guide

| Data | Possible choice | Reason |
|---|---|---|
| Product catalog | Daily/incremental batch | Changes relatively slowly |
| Orders | Incremental or CDC | Frequent changes; hourly reporting |
| Inventory events | CDC or streaming | Fresh stock visibility may matter |
| Warehouse reference data | Daily batch | Small and slowly changing |
| Historical rebuild | Full batch | Large bounded recovery task |

These are suggestions, not the only correct answers. Defend your choices.


# Phase 4 — design analytics

**Time: 20 minutes**

Include:

- Landing or raw-data area
- Validation and standardization
- Integration of order, inventory, and fulfillment data
- Analytical warehouse/OLAP store
- Historical preservation
- Reports or analytical outputs

Optional:

- MPP note for large analytical queries
- Curated sales and inventory subject areas
- Forecasting or recommendation consumer


# Analytical grain prompt

State what one row represents in at least one analytical dataset.

Examples:

```text
Sales fact:       one product within one completed order
Inventory event:  one stock change for one product at one warehouse
Fulfillment fact: one shipment for one order
```

Why it matters:

> If the grain is unclear, totals and joins may be wrong.


# Phase 5 — add quality and governance

**Time: 10 minutes**

Add notes for at least:

- One data-quality check
- One security control
- One metadata or ownership control
- One recovery requirement

Examples:

- Order total must not be negative
- Customer contact data is restricted
- Sales dataset has an owner and freshness label
- Failed movement can restart without duplicate orders


# Phase 6 — architecture review

**Time: 10 minutes**

Trace one order from beginning to end:

```text
mobile/web
   → order creation
   → stock reservation
   → warehouse processing
   → analytical movement
   → transformation
   → sales report
```

At every arrow, ask:

- What data moves?
- Which system owns it?
- When does it move?
- What happens if the step fails?


# Keep the diagram readable

Your final image should:

- Fit on one 16:9 slide
- Use readable text at normal presentation size
- Prefer 12–18 meaningful boxes over dozens of tiny boxes
- Avoid crossing arrows where possible
- Use alignment and spacing consistently
- Include a title and legend
- Distinguish operational and analytical boundaries
- Avoid unexplained abbreviations

Clarity is part of architecture.

# What earns a strong result?

A strong submission:

- Tells a complete data story
- Separates operational and analytical responsibilities
- Shows authoritative ownership
- Uses movement patterns appropriate to freshness needs
- Makes history available without burdening checkout
- Connects outputs to real consumers
- Labels assumptions and trade-offs
- Can be explained in under three minutes

The best diagram is not the one with the most boxes.

# Reference pattern — reveal after submission

```text
Customers ─► Mobile/Web ─► Order service ─► Orders OLTP
                                      │
Payment provider ◄────────────────────┤
                                      ├─► Inventory service ─► Inventory OLTP
                                      │
Warehouse staff ─► Warehouse app ─────┘

Orders OLTP ───── incremental/CDC ──┐
Inventory OLTP ── incremental/CDC ──┼─► raw landing
Product data ─────── daily batch ───┘       ↓
                                  validate + integrate
                                             ↓
                                   analytical warehouse
                                    ↙       ↓        ↘
                                reports  analysts  forecast
```

This is one valid pattern—not the only correct answer.

# Practice-session takeaway

```text
Enterprise data flow =
business events
  + clear system ownership
  + appropriate data movement
  + analytical preparation
  + useful consumers
  + visible trade-offs
```

If a diagram shows only technology boxes, it is incomplete.

If it shows how business evidence becomes trustworthy action, it is an architecture.
