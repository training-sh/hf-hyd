# AlturaWave Electronics

## Global Partner Data Business Case

**Case-study period:** 1 January 2025-31 December 2025  
**Audience:** Executive management and the newly hired Data Specialist  
**Document purpose:** Define the business problem before any solution is designed

> AlturaWave can manufacture world-class products, but it cannot yet see one trustworthy version of what its partners sold, stocked, and repaired.


# 1. Executive Scenario

AlturaWave Electronics is a fictional Singapore-headquartered consumer-electronics manufacturer with annual revenue of approximately **USD 4.8 billion**. It sells televisions, laptops, and monitors across five sales countries, primarily through independent retailers and distributors.

Partners periodically return sales, inventory, and warranty data. Each partner reports in its own format, terminology, currency, identifiers, timing, and level of detail. Management therefore receives regional spreadsheets that disagree and arrive too late for confident decisions.

The immediate business need is finite: create reliable visibility across **three inbound processes - sales, inventory snapshots, and warranty claims - for calendar year 2025**. No technical solution has yet been selected.

# 4. Sales Channels

| Channel | Commercial role | Typical reporting behavior |
|---|---|---|
| Large retail chain | High-volume store and online sales | Daily, detailed transactions |
| Regional retail outlets | Local market reach | Weekly consolidated files |
| Distributor | Resells to smaller outlets | Weekly invoices and warehouse stock |
| Export distributor | Cross-border regional distribution | Monthly invoice extracts |
| Online retail partner | Marketplace and fulfilled online orders | Daily event-oriented files |
| Refurbished seller | Sells graded returned or repaired products | Weekly sales and claims |

Channel comparisons are difficult because a distributor may report its sale to an outlet, while a retailer reports the final consumer sale. Management must not casually treat those events as identical.

# 6. Manufacturer-Owned Master Data

AlturaWave owns and governs:

- the nine official product IDs and SKUs,
- official names, models, category hierarchy, launch dates, list prices, and lifecycle status,
- the six official partner IDs, names, contracts, and expected reporting schedules,
- authorized partner-product mappings when known,
- corporate region and country definitions,
- approved failure-category vocabulary,
- monthly planning exchange rates.

This information is authoritative for AlturaWave, but partners do not consistently use it. Partner files remain operational evidence supplied by external organizations and are not automatically authoritative master data.


# 7. Partner-Owned Operational Data

Partners own the systems that create:

- checkout receipts, invoices, orders, and returns,
- partner-specific product and location codes,
- local selling prices and discounts,
- daily, weekly, or month-end stock positions,
- warranty intake, repair, replacement, and refund events,
- local geographic labels and timestamps.

AlturaWave can require a contractual feed but cannot assume identical operational processes. Corrections may arrive after initial submission, and a partner's understanding of "sale," "available stock," or "claim closed" may differ from another partner's.


# 8. Sales Reporting

A useful partner sale record may contain:

| Business content | Common partner variations |
|---|---|
| Transaction reference | `transaction_id`, `invoice_no`, `orderRef` |
| Product reference | `product_code`, `sku`, `material_number`, `item_id` |
| Event date/time | `sale_date`, `invoice_date`, `transaction_timestamp` |
| Selling location | store, outlet, warehouse, or virtual channel ID |
| Quantity | `qty`, `quantity`, `units_sold` |
| Price and discount | unit price, gross value, discount value or percentage |
| Net amount | `amount`, `net_sales`, `invoice_value` |
| Context | currency, channel, condition, city/state/country |

Some distributor records represent sell-in to an outlet, while retail and online records generally represent sell-through. Returns may arrive as negative quantities or separate transaction types.


# 9. Inventory Reporting

Partners may provide product identifier, location, snapshot date, quantity on hand, quantity reserved, and quantity available.

| Partner | Snapshot cadence | Important limitation |
|---|---|---|
| MegaMart USA | Daily close | Missing a day can hide a stock-out |
| BharatLink | Weekly Sunday | Cannot show intraworkweek depletion |
| EuroBridge | Month-end | Too coarse for weekly production action |
| NimbusCart | Daily | One virtual location; reserved stock changes rapidly |
| RenewLoop | Weekly | Stock split by refurbishment grade |
| GulfStar | Weekly | Store and distribution stock mixed in one file |

`available` is not universal: some partners calculate on-hand minus reserved; others exclude damaged, display, or in-transit units differently.


# 10. Warranty Reporting

Expected business content includes claim ID, serial number, partner product code, original sale date when known, claim date, problem category, resolution, warranty cost, partner, and geography.

Controlled incompleteness is expected:

- MegaMart sales omit serial numbers, although warranty claims often contain them.
- NimbusCart provides an original order reference but no store.
- RenewLoop claims include product condition and refurbishment grade.
- GulfStar may omit original sale date for walk-in claims.
- BharatLink sometimes records a reseller rather than the end-customer geography.

A claim without a matching submitted sale is not automatically invalid; it may reflect missing history, a cross-border purchase, a code mismatch, or delayed partner reporting.


# 11. Partner Data Formats - CSV Example

MegaMart sends daily UTF-8 CSV using US date notation and its own product code.

```csv
txn_id,product_code,sale_date,store_id,qty,unit_price,discount,net_sales,currency,condition,state
MM-78421,MM-TV55-882,07/14/2025,US034,1,749.00,50.00,699.00,USD,new,CA
MM-78422,MM-LAP16-416,07/14/2025,US017,1,1399.00,100.00,1299.00,USD,new,Texas
```

`CA` is a state code while `Texas` is a full state name. The date is `MM/DD/YYYY` even though other partners use different conventions.


# 11. Partner Data Formats - JSON Example

NimbusCart sends daily JSON with timestamps and order-line identifiers.

```json
{
  "orderRef": "NC-GB-55018",
  "transaction_timestamp": "2025-07-14T21:16:42+01:00",
  "item_id": "NIM-V55Q",
  "units_sold": 1,
  "grossAmount": 659.00,
  "promoAmount": 40.00,
  "amount": 619.00,
  "currency": "GBP",
  "condition": "NEW",
  "country": "GB"
}
```

The timestamp includes a time zone; the partner item ID is not an AlturaWave ID.


# 11. Partner Data Formats - XML Example

BharatLink sends a weekly XML invoice extract.

```xml
<sale>
  <invoice_no>BL-2025-11892</invoice_no>
  <material_number>89010291</material_number>
  <invoice_date>14/07/2025</invoice_date>
  <warehouse>BLR-W1</warehouse>
  <quantity>4</quantity>
  <invoice_value currency="INR">249600.00</invoice_value>
  <market>India-South</market>
</sale>
```

The quantity is distributor sell-in, and the date is `DD/MM/YYYY`.


# 11. Partner Data Formats - Parquet-Style Schema

EuroBridge supplies a typed month-end export extract.

```text
message eurobridge_sales {
  required binary document_no (STRING);
  required binary article_no (STRING);
  required int32 invoice_day (DATE);
  required int32 units;
  required int64 net_value_minor;
  required binary currency_code (STRING);
  optional binary destination_iso2 (STRING);
  optional binary hub_code (STRING);
}
```

`net_value_minor` is stored in cents, unlike decimal amounts in the other feeds.


# 11. Partner Data Formats - Avro-Style Schema

RenewLoop supplies weekly refurbished-product events.

```json
{
  "type": "record",
  "name": "RefurbSale",
  "fields": [
    {"name": "saleRef", "type": "string"},
    {"name": "sellerSku", "type": "string"},
    {"name": "soldOn", "type": {"type": "int", "logicalType": "date"}},
    {"name": "units", "type": "int"},
    {"name": "grade", "type": {"type": "enum", "name": "Grade", "symbols": ["A", "B", "C"]}},
    {"name": "netGBP", "type": "double"}
  ]
}
```

Refurbishment grade affects price and warranty expectations but does not exist in new-product feeds.

# 13. Semantic Differences Between Partners

| Concept | Variations and ambiguity |
|---|---|
| Quantity | `qty`, `quantity`, `units`, `units_sold`; returns may be negative or separately typed |
| Date | sale date, invoice date, timestamp; `YYYY-MM-DD`, `DD/MM/YYYY`, `MM/DD/YYYY` |
| Amount | gross, net, tax-inclusive, tax-exclusive, or minor currency units |
| Currency | USD, INR, EUR, GBP, AED; one GulfStar file occasionally omits it |
| Geography | country name, ISO-2 code, partner region, state code, or full state name |
| Location | store, warehouse, hub, virtual outlet; IDs only have meaning inside a partner |
| Product condition | omitted means new for some partners; explicit `NEW`, `A`, `B`, or `C` for others |
| Sale meaning | consumer checkout, online fulfillment, or distributor invoice |

File-format conversion solves syntax. It does not resolve these business meanings.


# 14. Data Quality Problems

The teaching dataset will contain a controlled set of issues, not universal corruption:

1. Twelve sales rows use unknown product codes.
2. A MegaMart file is retransmitted, creating duplicate transaction lines.
3. Nine GulfStar sales omit an outlet identifier.
4. Returns appear as negative quantities without an explicit return flag.
5. Seven records omit currency.
6. Two weekly files arrive 10-14 days late.
7. One BharatLink file switches unexpectedly to `YYYY/MM/DD`.
8. MegaMart changes a product code in September.
9. Three expected inventory snapshots are absent.
10. Thirty-four warranty claims have no matching submitted sale.
11. Four warranty claim IDs occur twice after correction.
12. EuroBridge reports two net values in euros rather than euro cents.

Most records remain correct so learners can distinguish normal processing from exception handling.

During the 2025 case period:

- **1 June:** GulfStar moves from the "Middle East Export" classification to "Middle East Direct Partner."
- **15 July:** GulfStar outlet DXB-03 moves from Dubai Central territory to Dubai South.
- **1 September:** MegaMart changes its code for VistaPro 55 from `MM-TV55-882` to `MM-TV55-NEW`.
- **31 October:** PulseBook 15 becomes discontinued but remains sellable from channel inventory.
- **15 December:** Horizon 34 enters end-of-life status while warranty obligations continue.

Management expects reports to explain performance in the appropriate historical business context without erasing the present view.


# 16. Management Business Questions - Sales

1. Which products generated the most units, revenue, and growth globally?
2. Which products perform best by country, region, partner, and channel?
3. Which partners are growing, stable, or declining month over month?
4. How do distributor sell-in patterns differ from retailer sell-through?
5. What share of sales is new versus refurbished?
6. How much discounting occurs by product and partner?
7. Are reported sales consistent with the official product portfolio and lifecycle status?


# 16. Management Business Questions - Inventory

8. Which product-location combinations are approaching stock-out?
9. Where is excess or aging inventory accumulating?
10. How many weeks of stock remain at the recent sales rate?
11. Which high-selling products have low reported channel inventory?
12. Which partners repeatedly miss inventory snapshots or report stale positions?

Comparisons must disclose differences in daily, weekly, and month-end reporting cadence.


# 16. Management Business Questions - Warranty

13. Which models have the highest reported claim rate relative to sales?
14. Are failure categories concentrated by country, partner, or product condition?
15. Which partners report unusually high repair, replacement, or refund rates?
16. What is the warranty cost by model and region?
17. How many claims cannot be associated with a submitted sale, and why?

Warranty rates must account for reporting completeness, sales timing, and the continued claim window of discontinued products.


# 16. Management Business Questions - Production Planning

18. Which products should receive near-term production priority?
19. Which regions show accelerating demand rather than a one-time distributor order?
20. Which products combine strong sales, low channel stock, and acceptable warranty performance?

These questions require careful interpretation. Partner data informs production planning, but it does not include factory capacity, component supply, procurement, or logistics constraints in this initial case.

# 18. Explicit Scope Boundaries

The initial case does **not** include:

- individual customer analytics or customer personally identifiable information,
- manufacturing plant operations,
- procurement or supplier management,
- transportation or logistics tracking,
- financial accounting,
- marketing campaigns,
- employee or HR data,
- IoT telemetry,
- real-time streaming,
- machine learning,
- implementation of the analytical platform.

Only partner sales, inventory snapshots, warranty claims, and the small manufacturer-owned reference set are in scope.


# 19. Assumptions

- Partners are contractually permitted to supply the described operational records.
- No end-customer names, addresses, emails, phone numbers, or payment details are provided.
- Manufacturer list price is a reference; partner selling prices legitimately vary.
- Monthly planning exchange rates are available, but the business must agree how they are used.
- Submitted records may be corrected or restated later.
- Location codes are only unique within a partner.
- Serial number is optional in sales and commonly present in warranty claims.
- A distributor invoice and consumer sale are both commercially valuable but different events.
- Production recommendations will be advisory because plant and supply constraints are out of scope.


# 20. Open Questions for the Data Specialist

1. Which management definitions of sale, return, net revenue, and available inventory are authoritative?
2. Should distributor sell-in and retail sell-through be compared, combined, or reported separately?
3. Which date determines the reporting period when event, invoice, and receipt dates differ?
4. What is the approved exchange-rate rule and reporting currency?
5. How should corrected and resent partner files be recognized?
6. Who owns product-code and location-code mapping approval?
7. How long may a feed be late before management is warned?
8. What confidence threshold is required before a warranty claim is associated with a sale?
9. How should missing inventory snapshots affect stock-out alerts?
10. Which historical classification should each report present?
11. What reconciliation evidence will finance and partner managers accept?
12. Who may view partner-level prices, discounts, and warranty costs?


# Data Specialist Assignment

AlturaWave management asks you to establish a reliable analytical platform that combines inconsistent partner sales, inventory, and warranty feeds with the company's official product and partner information.

The platform must give management a clear, explainable view across products, partners, channels, regions, and time while preserving important differences in business meaning. It must identify unmapped codes, missing submissions, duplicates, late corrections, and other exceptions rather than silently hiding them.

Begin by validating the business definitions, ownership, mappings, reporting cadence, quality expectations, security rules, and acceptance criteria with stakeholders. Recommend a delivery approach only after the unresolved questions are answered.

Success means that authorized decision-makers can trust the reported sales, inventory, and warranty results, understand their limitations, trace discrepancies to their origin, and use the evidence to improve channel management and production priorities.
