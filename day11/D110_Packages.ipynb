# D110 — Python Packages

A **module** is usually one `.py` file. A **package** is an importable directory that groups modules and often subpackages.

A package can contain `__init__.py`, modules, subpackages, data files, and compiled `.pyd` extensions. Installation also creates a `.dist-info` directory containing version, dependency, and file metadata.

We will install `requests` normally and `python-dateutil` in a custom directory.

## 3. Install into a custom directory

`--target` installs into a chosen directory instead of normal `site-packages`. For this teaching exercise we use the disposable location `C:\tmp\python-packages`.

````python
from pathlib import Path

custom_packages = Path(r"C:\tmp\python-packages")
custom_packages.mkdir(parents=True, exist_ok=True)
print(custom_packages)
````

````python
%pip install --target "C:\tmp\python-packages" python-dateutil
````

Command Prompt equivalent:

```bat
C:\Users\GOPALAKRISHNANSUBRAM\dataeng\Scripts\activate.bat
python -m pip install --target "C:\tmp\python-packages" python-dateutil
```

`C:\tmp\python-packages` is only a chosen teaching location; it is not a special Python directory.

## 4. Where Python looks for packages

`import` searches each directory in `sys.path`, in order. It normally includes the script/notebook location, the standard library, the active environment's `site-packages`, and any entries supplied by `PYTHONPATH`.

````python
for number, folder in enumerate(sys.path):
    print(f"{number:2}: {folder}")
````

A `--target` directory is not automatically added. Add the directory that **contains** the package folder. Here, `custom-packages` contains the `dateutil` folder.

````python
custom_path = str(custom_packages)
if custom_path not in sys.path:
    sys.path.insert(0, custom_path)  # current kernel only

import dateutil
from dateutil.parser import parse

print("version:", dateutil.__version__)
print("loaded from:", dateutil.__file__)
print(parse("29 July 2026 10:30 PM"))
````

## 5. `PYTHONPATH` on Windows

`PYTHONPATH` is optional. It adds import roots to `sys.path` when Python starts. Separate multiple Windows paths with `;`.

For the current Command Prompt window:

```bat
set "PYTHONPATH=C:\tmp\python-packages;%PYTHONPATH%"
python -c "import dateutil; print(dateutil.__file__)"
```

Save it for the current Windows user:

```bat
setx PYTHONPATH "C:\tmp\python-packages;%PYTHONPATH%"
```

`set` affects only the current window. `setx` affects new Command Prompt and Jupyter processes; restart them afterward. Preserve any existing required entries, joined with `;`.

## 6. `PATH`, `PYTHONPATH`, and `Lib`

- `PATH` helps Windows find executables such as `python.exe`; it does not control imports.
- `PYTHONPATH` adds directories to Python's module search path.
- `Lib` contains Python's standard library.
- `Lib\site-packages` contains third-party packages for that environment.

Prefer a virtual environment's `site-packages` for normal projects. Use `--target` when a custom bundle or teaching example requires it.
