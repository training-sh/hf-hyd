Goal: compare vague and specific prompts, system instructions, generation controls, streaming and conversation history. Allow 35 minutes.

Run cells from top to bottom. Complete notebook 00 first. Each notebook starts with fresh state.


```python
from google.genai import types
from course_config import PROJECT_ID, MODEL_ID, EMBEDDING_LOCATION, make_client
client = make_client()
print('Project:', PROJECT_ID, '| Model:', MODEL_ID)
```

A language model predicts tokens conditioned on context. Good prompts specify the task, audience, constraints and examples. Temperature changes sampling; temperature zero is not a guarantee of identical results.


```python
prompts = [
    'Explain BigQuery.',
    'You are teaching a Python beginner. Explain BigQuery in 80 words, include a warehouse analogy and one SQL example.'
]
for prompt in prompts:
    result = client.models.generate_content(model=MODEL_ID, contents=prompt,
        config=types.GenerateContentConfig(temperature=0.3, max_output_tokens=1200))
    print('PROMPT:', prompt, '\nANSWER:', result.text, '\n')
```

```python
chat = client.chats.create(model=MODEL_ID, config=types.GenerateContentConfig(
    system_instruction='Teach a beginner. Use short explanations and concrete Python examples.',
    max_output_tokens=1200))
print(chat.send_message('What is an embedding?').text)
print(chat.send_message('How is that different from a token?').text)
```

The chat object maintains context in this Python process. It is not permanent model memory. Longer histories increase input token usage.


```python
for part in client.models.generate_content_stream(model=MODEL_ID,
        contents='Give three practical uses for Gemini in data engineering.',
        config=types.GenerateContentConfig(max_output_tokens=1200)):
    if part.text:
        print(part.text, end='')
```
