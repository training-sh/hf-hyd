Embeddings represent meaning as numeric vectors; they are not token IDs. Similarity compares vectors from the same model and dimensionality. Cosine similarity measures angle, not a probability of correctness.


```python
import numpy as np
from lab_utils import normalize
vectors = normalize([[1, 0], [0.9, 0.1], [0, 1]])
print('Toy cosine similarity matrix:\n', vectors @ vectors.T)
```

```python
from lab_utils import load_documents, embed_texts
embedding_client = make_client(EMBEDDING_LOCATION)
documents = load_documents()
matrix = embed_texts(embedding_client, [d['text'] for d in documents], 'RETRIEVAL_DOCUMENT')
print('Rows = documents, columns = dimensions:', matrix.shape)
query = 'Can I carry unused vacation into next year?'
q = embed_texts(embedding_client, [query], 'RETRIEVAL_QUERY')[0]
scores = matrix @ q
for i in np.argsort(-scores)[:3]:
    print(documents[i]['id'], round(float(scores[i]), 3), documents[i]['text'])
```

Here NumPy performs exact search over a tiny matrix. A vector database adds indexing, persistence, metadata filters and operational features. Approximate nearest-neighbor search trades some recall for scale. Re-embed documents if you change the embedding model or dimensions.
