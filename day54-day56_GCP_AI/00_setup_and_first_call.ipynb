```python
from google.genai import types
from course_config import PROJECT_ID, MODEL_ID, EMBEDDING_LOCATION, make_client
client = make_client()
print('Project:', PROJECT_ID, '| Model:', MODEL_ID)
```

## Authentication preflight
Application Default Credentials (ADC) identify the caller. The project identifies the resource/quota scope. Logging into the browser alone does not authenticate a local Python kernel. Do not enter the sandbox password into notebook code.


```python
import google.auth
credentials, detected_project = google.auth.default(scopes=['https://www.googleapis.com/auth/cloud-platform'])
print('Credential type:', type(credentials).__name__)
print('ADC project:', detected_project)
# Never print credential objects, access tokens, or refresh tokens.
```

## First network call
This consumes sandbox model quota. Expected: a short explanation of generative AI; exact wording varies.


```python
response = client.models.generate_content(model=MODEL_ID,
    contents='Explain generative AI to a Python beginner in three sentences.',
    config=types.GenerateContentConfig(max_output_tokens=1000))
print(response.text)
print(response.usage_metadata)
```

## If the call fails
- Credentials missing or 401: complete ADC login for the sandbox account in README.
- 403: check the sandbox project, Vertex AI API, and lab-provided permissions. Some labs block models or regions.
- 404: check MODEL_ID and LOCATION in course_config.py against models available to the lab.
- 429: pause and retry once; repeated calls consume quota and may not resolve capacity limits.
- Empty text: inspect response.candidates for finish reasons; a response may be blocked or exhaust its output budget.

No endpoint deployment is needed for these publisher-model calls.
