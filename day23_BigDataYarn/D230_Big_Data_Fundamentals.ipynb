# Big Data Fundamentals

### From growing data to useful decisions

**Today:** evolution · need · 5Vs · core components · business use cases

# Why do we need Big Data?

Traditional systems work well when data is structured, predictable, and moderate in size.

Big Data becomes relevant when we need to:

- store more data than one machine can handle comfortably
- process events as they arrive
- combine tables, logs, text, images, and sensor data
- serve many users and workloads reliably
- turn raw data into timely decisions

> Big Data is not only about size. It is about solving data problems at scale.

# Evolution of Big Data

| Stage | Typical data | Main approach |
|---|---|---|
| **Operational era** | Business transactions | Relational databases |
| **Web era** | Clicks, searches, server logs | Distributed storage and batch processing |
| **Cloud era** | Apps, mobile, IoT, media | Elastic storage and compute |
| **Real-time & AI era** | Event streams and unstructured data | Streaming, lakehouse, ML platforms |

The key shift: **from recording what happened to predicting and responding to what is happening.**

<small>Hadoop was an important step in distributed batch processing, but it is one part of this larger evolution.</small>

# The 5Vs of Big Data

| V | The question it raises |
|---|---|
| **Volume** | How much data must we handle? |
| **Velocity** | How quickly does data arrive and need a response? |
| **Variety** | How many forms and sources of data are involved? |
| **Veracity** | Can we trust the data? |
| **Value** | Does the data improve an outcome? |

> The 5Vs describe the challenge. They also guide the architecture.

# Volume and Velocity

## Volume — the scale of data
- Millions of transactions, events, files, or device readings
- Calls for scalable storage, partitioning, and parallel processing
- Example: years of order history plus every customer click

## Velocity — the speed of data
- Data may arrive continuously and lose value quickly
- Calls for event ingestion, stream processing, and low-latency serving
- Example: flagging a suspicious card payment before approval

**Think:** Volume asks, “Can we hold and process it?” Velocity asks, “Can we act in time?”

# Variety, Veracity, and Value

## Variety — different shapes
Tables, JSON events, text, audio, images, and sensor readings need different ways to store and process them.

## Veracity — data we can trust
Missing values, duplicates, late events, bias, and inconsistent definitions can produce the wrong answer. Quality checks and governance matter.

## Value — a useful outcome
More data is not automatically better. Value appears when data reduces cost, increases revenue, lowers risk, or improves an experience.

**Example:** Customer reviews have variety; fake reviews affect veracity; better product decisions create value.

# Components of a Big Data System

**Sources** → **Ingestion** → **Storage** → **Processing** → **Serving** → **Use**

- **Sources:** applications, databases, APIs, devices, files
- **Ingestion:** collect data in batches or as event streams
- **Storage:** object storage, data lakes, warehouses, lakehouses
- **Processing:** clean, join, aggregate, and enrich data
- **Serving:** make trusted data fast and easy to access
- **Use:** dashboards, alerts, analytics, and machine learning

**Across every layer:** security · data quality · metadata · governance · monitoring

# Business Use Cases

| Business need | Data used | Outcome |
|---|---|---|
| Personalization | clicks, searches, purchases | relevant recommendations |
| Fraud detection | payment and device events | block risky activity quickly |
| Predictive maintenance | sensor readings and service logs | reduce equipment downtime |
| Supply-chain planning | orders, inventory, location, weather | fewer delays and stock-outs |
| Customer experience | calls, chats, reviews, usage | find issues and reduce churn |
| Dynamic operations | demand, traffic, capacity | better pricing and allocation |

# From Use Case to Data Solution

Start with the decision, not the technology.

1. **Decision:** What action should improve?
2. **Data:** Which signals support that decision?
3. **Timing:** Is the answer needed in seconds, hours, or days?
4. **Scale:** How much data and how many users?
5. **Trust:** How will quality, privacy, and access be controlled?
6. **Measure:** Which business metric proves value?

> A successful Big Data system connects scale to a measurable business result.

# Key Takeaways

- Big Data is about handling data complexity at scale.
- The 5Vs help us understand the problem.
- Storage and processing are core, but ingestion, serving, and governance complete the system.
- Architecture should follow the business use case.
- Data creates value only when it improves a decision or action.
