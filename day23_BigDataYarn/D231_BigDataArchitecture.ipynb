# Big Data Architecture

### How data moves from source to business use

**Data Sources · Storage Layer · Processing Layer · Consumption Layer**

# The General Architecture

## Sources → Ingestion → Storage → Processing → Consumption

| Layer | Main responsibility |
|---|---|
| **Data Sources** | Produce data |
| **Ingestion** | Collect and move data |
| **Storage** | Keep data reliably at scale |
| **Processing** | Clean, combine, and transform data |
| **Consumption** | Deliver data for decisions and applications |

**Across all layers:** security · governance · metadata · quality · monitoring

# Data Sources

Big Data systems collect data from many operational environments.

- **Business systems:** orders, payments, inventory, CRM
- **Applications:** clicks, searches, logs, API events
- **Devices:** sensors, machines, vehicles, mobile devices
- **Files and media:** CSV, JSON, documents, images, audio
- **External sources:** partners, public data, market feeds

Source data may be **structured, semi-structured, or unstructured**.

> First understand who produces the data, how often it arrives, and how reliable it is.

# Ingestion: Connecting Sources to the Platform

## Batch ingestion
- Moves data on a schedule
- Suitable for files, database extracts, and daily reports
- Simple and efficient when immediate results are not required

## Stream ingestion
- Moves events continuously
- Suitable for clicks, payments, logs, and sensors
- Supports alerts and near-real-time decisions

**Design for:** retries, duplicate events, late data, schema changes, and traffic spikes.

# Storage Layer

The storage layer keeps data durable, scalable, and accessible.

| Storage approach | Best suited for |
|---|---|
| **Data lake / object storage** | raw and varied data at large scale |
| **Data warehouse** | governed, structured analytical data |
| **Lakehouse** | lake-scale storage with table management |
| **Operational store** | fast lookups for applications |

Organize data by business domain, date, or other common access pattern.

> Storage is not just a destination. Its layout affects cost, speed, and trust.

# Organizing Data in Storage

A common pattern separates data by level of preparation.

1. **Raw:** data preserved close to its original form
2. **Cleaned:** validated, standardized, and deduplicated data
3. **Curated:** business-ready tables and aggregates

Why keep these stages?

- raw data supports replay and investigation
- cleaned data gives teams consistent inputs
- curated data makes common analysis faster

**Also plan for:** file format, partitioning, retention, access control, and metadata.

# Processing Layer

Processing converts stored or incoming data into useful information.

- validate fields and business rules
- remove duplicates and handle missing data
- join data from multiple sources
- calculate metrics and aggregates
- enrich data with reference information
- prepare features for machine learning

Processing can run on **one record, a time window, a partition, or the complete data set**.

# Batch or Stream Processing?

| Question | Batch | Stream |
|---|---|---|
| When does it run? | on a schedule | continuously |
| Data handled | bounded collection | ongoing events |
| Typical latency | minutes to hours | milliseconds to minutes |
| Typical use | daily sales report | payment fraud alert |
| Main strength | simpler large-scale computation | timely response |

Many platforms use both. Choose according to the required business response time.

> Real time adds complexity. Use it when the value of the result falls quickly with time.

# Consumption Layer

The consumption layer makes processed data useful.

- **Dashboards:** trends, KPIs, and operational status
- **Reports:** scheduled and governed business information
- **Ad hoc analysis:** exploration using SQL or notebooks
- **Applications and APIs:** data delivered inside products
- **Alerts:** action triggered by a condition or event
- **Machine learning:** training data and model predictions

Consumption needs differ: freshness, response time, detail, concurrency, and access control.

> One storage format rarely serves every consumer efficiently.

# Example: Retail Order Analytics

| Layer | What happens |
|---|---|
| **Sources** | orders, payments, product catalog, click events |
| **Ingestion** | orders stream continuously; catalog loads daily |
| **Storage** | raw events retained; clean order tables maintained |
| **Processing** | validate orders, join products, calculate sales |
| **Consumption** | sales dashboard, low-stock alert, recommendations |

The same data platform supports different response times:

- low-stock alert: minutes
- sales dashboard: hourly
- finance report: daily
- long-term demand model: weekly or monthly

# Architecture Decisions

Before selecting technologies, answer these questions:

1. What business decision or action must the system support?
2. Which sources are required, and can they be trusted?
3. How much data arrives, and how quickly?
4. How fresh must the result be?
5. Who will consume the data, and in what form?
6. What are the security, privacy, retention, and recovery needs?
7. How will cost, reliability, and data quality be monitored?

> Good architecture is driven by requirements, not by the number of tools.

# Key Takeaways

- Sources define the shape, speed, and quality of incoming data.
- Ingestion moves data safely in batches or streams.
- Storage keeps raw and prepared data at the required scale.
- Processing turns data into trusted business information.
- Consumption delivers the result in a useful form and time.
- Security, quality, governance, and monitoring span every layer.

## Final thought
**Trace every architecture choice back to a business need.**
