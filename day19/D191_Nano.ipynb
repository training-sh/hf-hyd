# D191 — Nano Text Editor Basics on WSL/Ubuntu

Nano is a small terminal text editor suitable for configuration files, scripts, notes, and quick changes. This lab uses direct, copy-paste-friendly exercises inside `$HOME/nano_lab`.

> Run commands in the Ubuntu/WSL terminal. Nano itself is interactive: commands shown with `Ctrl` or `Alt` are keyboard shortcuts, not text to type into the file.

## 1. Check or install Nano

```bash
nano --version
command -v nano
```

If Nano is not installed:

```bash
sudo apt update
sudo apt install nano
```

Open Nano's help or manual from the shell:

```bash
nano --help
man nano
```

Press `q` to leave the manual.

## 2. Create a safe practice directory

```bash
mkdir -p "$HOME/nano_lab"
cd "$HOME/nano_lab"
pwd
printf 'Linux\nPython\nData Engineering\n' > topics.txt
ls -l topics.txt
```

Open the existing file:

```bash
nano topics.txt
```

Nano fills the terminal. The filename appears near the top, the file contents occupy the middle, status messages appear near the bottom, and common shortcuts appear on the final two lines.

## 3. Read Nano's shortcut notation

Nano displays shortcuts such as `^O` and `M-U`:

| Nano notation | Meaning | Typical keyboard action |
|---|---|---|
| `^` | Control key | `^O` means `Ctrl+O` |
| `M-` | Meta key | Usually `Alt`; `M-U` means `Alt+U` |
| `Space` | Spacebar | Press the spacebar |
| `Enter` | Enter/Return | Confirm an action |

On some terminals, an `Alt` shortcut may not be delivered correctly. Press `Esc`, release it, and then press the shortcut letter as an alternative. Nano shows more shortcuts with `Ctrl+G`; use `Ctrl+X` to close the help screen.

## 4. Enter text and move around

Open a new file:

```bash
cd "$HOME/nano_lab"
nano notes.txt
```

Type these lines normally:

```text
Nano is a terminal editor.
It is useful for quick changes.
This is our practice file.
```

Navigation shortcuts:

| Action | Shortcut |
|---|---|
| Move one character/line | Arrow keys |
| Beginning of line | `Ctrl+A` |
| End of line | `Ctrl+E` |
| Previous page | `Ctrl+Y` |
| Next page | `Ctrl+V` |
| First line | `Alt+\` |
| Last line | `Alt+/` |
| Go to line and column | `Ctrl+_`, enter values, then `Enter` |

`Home`, `End`, `Page Up`, and `Page Down` commonly work too.

## 5. Save with Write Out

In Nano, saving is called **Write Out**.

1. Press `Ctrl+O`.
2. Nano displays `File Name to Write`.
3. Keep or edit the filename.
4. Press `Enter` to save.

A status message reports how many lines were written. Continue editing after saving.

To save a copy under another name, press `Ctrl+O`, replace the displayed filename with `notes-copy.txt`, and press `Enter`. This changes Nano's current filename to the new name.

## 6. Exit Nano safely

Press `Ctrl+X` to exit.

- If nothing changed since the last save, Nano exits immediately.
- If there are unsaved changes, Nano asks `Save modified buffer?`.
- Press `Y` to save, `N` to discard the unsaved changes, or `Ctrl+C` to cancel the exit.
- After `Y`, confirm the filename with `Enter`.

Verify from the shell:

```bash
cd "$HOME/nano_lab"
ls -l
cat notes.txt
```

Be careful with `N`: discarded edits generally cannot be recovered.

## 7. Delete, cut, copy, and paste text

Common editing shortcuts:

| Action | Shortcut |
|---|---|
| Delete character under cursor | `Delete` or `Ctrl+D` |
| Backspace | `Backspace` or `Ctrl+H` |
| Cut current line | `Ctrl+K` |
| Paste Nano's cut buffer | `Ctrl+U` |
| Undo | `Alt+U` |
| Redo | `Alt+E` |

To move a line, place the cursor on it, press `Ctrl+K`, move elsewhere, then press `Ctrl+U`.

Nano's `Ctrl+K` and `Ctrl+U` use Nano's internal cut buffer. They are not the same as the Windows clipboard shortcuts.

## 8. Select, copy, cut, and paste a block

1. Move to the start of the desired text.
2. Press `Alt+A` to set the mark and begin selection.
3. Move the cursor to highlight text.
4. Press `Alt+6` to copy the selection, or `Ctrl+K` to cut it.
5. Move to the destination.
6. Press `Ctrl+U` to paste.

Press `Alt+A` again to cancel an active selection.

For the Windows clipboard in Windows Terminal, use the terminal's configured shortcuts—commonly `Ctrl+Shift+C` to copy and `Ctrl+Shift+V` to paste—or right-click. These are handled by the terminal, not by Nano.

## 9. Search and replace

Open the practice file:

```bash
nano "$HOME/nano_lab/notes.txt"
```

Search:

1. Press `Ctrl+W`.
2. Type the search text.
3. Press `Enter`.
4. Press `Alt+W` to find the next occurrence.

Replace:

1. Press `Ctrl+\`.
2. Enter the text to find and press `Enter`.
3. Enter the replacement and press `Enter`.
4. Press `Y` for one replacement, `N` to skip, `A` for all, or `Ctrl+C` to cancel.

Use Replace All carefully. Review the file before saving.

## 10. Open at a line and display line numbers

Open directly at line 3:

```bash
nano +3 "$HOME/nano_lab/topics.txt"
```

Open at line 3, column 2:

```bash
nano +3,2 "$HOME/nano_lab/topics.txt"
```

Display line numbers for this session:

```bash
nano -l "$HOME/nano_lab/topics.txt"
```

Inside Nano, `Alt+N` toggles line numbers. `Ctrl+C` normally reports the current cursor line, column, and character position in the status area.

## 11. Insert another file and work with multiple files

Inside Nano, `Ctrl+R` reads another file into the current buffer at the cursor position. Nano asks for the filename; type `$HOME/nano_lab/topics.txt` and press `Enter`.

Open several files as separate buffers from the shell:

```bash
nano "$HOME/nano_lab/notes.txt" "$HOME/nano_lab/topics.txt"
```

Switch buffers with:

- `Alt+<` for the previous buffer.
- `Alt+>` for the next buffer.

When exiting, Nano asks about unsaved changes in each modified buffer.

## 12. Useful command-line options

```bash
nano -l file.txt             # show line numbers
nano -c file.txt             # continuously show cursor position
nano -m file.txt             # enable mouse support
nano -B file.txt             # create a backup when saving
nano -i file.txt             # auto-indent new lines
nano -w file.txt             # do not hard-wrap long lines
nano -T 4 file.txt           # display tab width as four columns
nano -v file.txt             # view-only mode
nano --help                  # list available options
```

Option availability can vary slightly by Nano version. Use `nano --help` on the WSL installation as the authority.

View-only mode discourages editing but is not an access-control mechanism. Linux file permissions determine whether a file can actually be changed.

## 13. Edit shell scripts

Create a script:

```bash
cd "$HOME/nano_lab"
nano hello.sh
```

Enter:

```bash
#!/usr/bin/env bash
set -euo pipefail
printf 'Hello, %s\n' "$USER"
```

Save with `Ctrl+O`, `Enter`; exit with `Ctrl+X`. Then validate, make executable, and run:

```bash
bash -n "$HOME/nano_lab/hello.sh"
chmod 750 "$HOME/nano_lab/hello.sh"
"$HOME/nano_lab/hello.sh"
```

Nano edits text; it does not automatically grant execute permission.

## 14. Configure Nano with `.nanorc`

User-specific settings belong in `$HOME/.nanorc`. Back up an existing file first:

```bash
test ! -f "$HOME/.nanorc" || cp -a "$HOME/.nanorc" "$HOME/.nanorc.backup.$(date +%Y%m%d_%H%M%S)"
nano "$HOME/.nanorc"
```

Example configuration:

```text
set linenumbers
set mouse
set tabsize 4
set tabstospaces
set autoindent
set indicator
```

Save and reopen Nano. If Nano reports an unknown option, check `man nanorc` because supported settings depend on the installed version.

To test without reading configuration:

```bash
nano -I "$HOME/nano_lab/notes.txt"
```

## 15. Edit protected configuration files safely

Normal files in your home directory do not need `sudo`. For a root-owned system file, prefer `sudoedit`:

```bash
sudoedit /path/to/system-file
```

Choose Nano as the editor for one command:

```bash
SUDO_EDITOR=nano sudoedit /path/to/system-file
```

`sudoedit` copies the protected file to a temporary file, opens the editor as the normal user, and writes it back with elevated permission after editing. It is generally safer than running the entire editor as root.

Before editing a real configuration file:

- Read its documentation.
- Create an appropriate backup.
- Validate syntax with the service's validation command.
- Change only what the task requires.
- Never experiment directly on an important production configuration.
