# What is data?

**Data is a recorded representation of something that happened, exists, or was observed.**

| Reality | Recorded as data |
|---|---|
| A customer places an order | Order ID, item, amount, timestamp |
| A package moves | Location scan, status, time |
| A person visits a page | Page, device, session, time |
| A sensor measures heat | Temperature, unit, timestamp |

Data is not reality itself. It is a **selective description** of reality.


# Data needs context

The value `42` alone is ambiguous.

- 42 **what**?
- Measured **where**?
- Recorded **when**?
- Produced **by whom or what**?
- Using **which definition and unit**?

`42` becomes meaningful as:

> `42 °C`, measured at Warehouse 3, by Sensor A17, at 14:05 UTC.

**Metadata** is data that explains data: names, types, units, ownership, meaning, and origin.


# From signals to decisions

```text
REAL-WORLD EVENT
       ↓ recorded
      DATA
       ↓ organized + interpreted
  INFORMATION
       ↓ combined with experience
    INSIGHT
       ↓ acted upon
    DECISION
```

Data becomes valuable only when people can understand it, trust it, and use it.

> **Think–pair–share (3 min):** Name one decision in your work or daily life that depends on data.


# Common shapes of data

| Shape | Description | Examples |
|---|---|---|
| **Structured** | Fixed fields and types | Orders table, account records |
| **Semi-structured** | Flexible labels and nesting | JSON events, application logs |
| **Unstructured** | No fixed row-and-column model | Emails, PDFs, images, audio |

The shape affects how easily data can be validated, combined, searched, and analyzed.

> “Unstructured” does not mean “without meaning.” It means the structure is not expressed as a rigid data model.


# Records, fields, and relationships

An order can be described as a record:

| order_id | customer_id | ordered_at | total | status |
|---|---|---|---:|---|
| O-1042 | C-208 | 2026-07-24 10:31 | 1499.00 | Shipped |

- A **field** describes one property
- A **record** groups properties about one thing or event
- A **relationship** connects records, such as an order to a customer
- An **identifier** distinguishes one record from another

Good identifiers make reliable connections possible.


# Two worlds of organizational data

<table style="font-size: 1.05rem;">
<tr><th style="background:#dbeafe;">Operational world</th><th style="background:#dcfce7;">Analytical world</th></tr>
<tr><td>Runs the business now</td><td>Understands the business over time</td></tr>
<tr><td>“Place this order”</td><td>“Why did sales fall last month?”</td></tr>
<tr><td>Fast individual actions</td><td>Large scans and comparisons</td></tr>
<tr><td>Current state matters most</td><td>History and trends matter most</td></tr>
</table>

These worlds serve different needs. Neither is “better.”


# Operational data

Operational data supports day-to-day processes:

- Create or cancel an order
- Update an address
- Reserve one unit of inventory
- Record a payment
- Assign a support ticket

Typical properties:

- Many small, frequent changes
- Very fast responses
- Strong correctness for each transaction
- Mostly current state
- Designed around an application or workflow


# Analytical data

Analytical data supports questions across many events:

- Revenue by region for the last 12 months
- Average delivery delay by carrier
- Repeat-purchase rate by customer segment
- Refund trends by product category

Typical properties:

- Reads many records at once
- Combines multiple business areas
- Preserves history
- Uses consistent business definitions
- Designed around comparison and aggregation


# Same data, different questions

| Operational question | Analytical question |
|---|---|
| Did order O-1042 pay successfully? | What percentage of orders failed payment this quarter? |
| Is item P-77 in stock now? | Which products had the most stockouts? |
| What is this customer’s address? | Which regions have the highest customer growth? |
| Has this package shipped? | How has delivery time changed by carrier? |

> Operational systems answer **“What is happening to this item?”**  
> Analytical systems answer **“What pattern appears across many items?”**


# Quick classification

Decide: **operational**, **analytical**, or **both**?

1. Authorize a card payment
2. Show five years of monthly revenue
3. Display a passenger’s current seat assignment
4. Predict which flights may be delayed
5. Update today’s product price
6. Compare campaign performance by channel

# Where does data come from?

```text
People ─────────┐
Applications ───┤
Machines ───────┼──► Organizational data
Partners ───────┤
Public sources ─┘
```

Every source has its own purpose, format, timing, quality, and owner.

The source usually creates data to perform **its own job**, not to make downstream analytics easy.


# Human-generated data

People create data intentionally:

- Registration and order forms
- Surveys and ratings
- Support notes
- Spreadsheets
- Documents and emails

Common challenges:

- Typing errors and inconsistent conventions
- Missing optional fields
- Personal interpretations of a category
- Copy-and-paste duplication

> Humans understand context naturally; systems need that context made explicit.


# Machine-generated data

Systems generate data as a by-product of activity:

- Web and mobile click events
- Application logs
- Sensors and connected devices
- GPS location signals
- Machine health metrics
- Security and access events

Machine-generated does **not** guarantee accuracy. Devices fail, clocks drift, events repeat, and software changes.


# Internal and external sources

| Internal | External |
|---|---|
| Sales and orders | Demographic data |
| Inventory | Weather |
| Finance | Market prices |
| Human resources | Partner feeds |
| Product usage | Government/open data |

External data can enrich internal facts:

> Delivery records + weather data → understand weather-related delays.

But ownership, licensing, reliability, and update frequency must be understood.


# Source systems have boundaries

At QuickCart:

- The order system knows purchases
- The payment provider knows authorizations
- The warehouse system knows stock movement
- The carrier knows delivery scans
- The support platform knows complaints

No single system sees the whole customer journey.

**Data engineering connects partial views without erasing their meaning.**


# The data engineering lifecycle

```text
SOURCES
   ↓
INGESTION ──► STORAGE ──► TRANSFORMATION ──► SERVING
                 ▲               │               ↓
                 └──── quality, governance ── CONSUMERS
```

This is a **lifecycle**, not merely a one-way pipeline:

- Definitions evolve
- Sources change
- Consumers provide feedback
- Quality must be monitored continuously
- Old data may need to be reprocessed


# Step 1 — Generation

Data is created when an event occurs or a state changes.

For order `O-1042`:

1. Customer submits the cart
2. Order record is created
3. Payment is authorized
4. Inventory is reserved
5. Shipment is scanned
6. Delivery is confirmed

Each step may happen in a different system and at a different time.


# Step 2 — Ingestion

**Ingestion** brings data from a source into a destination where it can be processed.

Two broad patterns:

| Batch | Continuous |
|---|---|
| Moves a collection periodically | Moves events soon after they occur |
| Hourly, daily, weekly | Seconds or milliseconds |
| Simpler when immediacy is unnecessary | Useful when rapid reaction matters |

Choose based on business need—not excitement.

> If tomorrow morning is soon enough, “real time” may add cost without adding value.


# Data can arrive imperfectly

Movement across systems introduces difficult realities:

- The same event may arrive twice
- Events may arrive out of order
- A late event may belong to yesterday
- A source may resend old records
- Part of a batch may fail
- Fields may appear, disappear, or change meaning

Reliable movement requires answering:

> What makes an event unique, complete, timely, and safe to process again?


# Step 3 — Storage

Storage provides a durable home for data.

Different layers may keep:

- **Raw data** — close to the source; useful for recovery and audit
- **Cleaned data** — standardized and validated
- **Curated data** — organized around trusted business concepts
- **Serving data** — shaped for a specific report, model, or application

Keeping layers separates preservation from interpretation.


# Step 4 — Transformation

Transformation changes data into a more useful form:

- Standardize dates, currencies, and units
- Correct or flag invalid values
- Remove duplicate events
- Connect orders to customers and products
- Calculate revenue, margin, and delivery time
- Apply agreed business definitions

Transformation is where raw records become shared business meaning.


# Step 5 — Serving and consumption

Serving makes prepared data accessible to consumers:

- Dashboards and reports
- Exploratory analysis
- Alerts
- Machine-learning features
- Regulatory extracts
- Data-powered product experiences

The lifecycle succeeds only when the delivered data is:

**understandable · trustworthy · timely · appropriate for the decision**


# Follow one order end to end

```text
Checkout       Payment       Warehouse       Carrier
   │              │              │              │
   └──────────── source records ────────────────┘
                          ↓
                       ingest
                          ↓
                  preserve raw events
                          ↓
          clean + join + apply definitions
                          ↓
              “daily delivery performance”
                          ↓
       analyst → manager → operational action
```

One dashboard number may depend on many systems and many transformations.


# Why must data be cleaned?

Real data reflects real-world messiness:

- People make mistakes
- Systems use different conventions
- Networks retry messages
- Definitions change
- Devices malfunction
- Business processes have exceptions

Cleaning does not mean making data look perfect.

**It means making quality problems visible and treating them consistently for an intended use.**


# Dimensions of data quality

| Dimension | Question |
|---|---|
| **Accuracy** | Does it reflect reality? |
| **Completeness** | Are required values present? |
| **Consistency** | Do systems agree? |
| **Validity** | Does it follow allowed rules? |
| **Uniqueness** | Is the same fact recorded once? |
| **Timeliness** | Is it available when needed? |

Quality is always relative to a purpose. A one-day delay may be fine for monthly reporting and unacceptable for fraud detection.


# A messy order

```text
order_id:     O-1042
country:      India
amount:       "1,499"
currency:     [missing]
ordered_at:   07/08/26 10:31
customer_id:  C-208
```

Questions before analysis:

- Is the date 7 August or 8 July?
- Is the amount numeric?
- Can currency be inferred safely?
- Is the country value standardized?
- Has this order already appeared?

> Cleaning requires rules—and rules require business knowledge.


# Cleaning is not rewriting history

Responsible handling preserves distinctions:

- **Raw value:** what the source actually sent
- **Standardized value:** a consistent representation
- **Derived value:** calculated from other data
- **Correction:** an approved change with traceability
- **Exclusion:** intentionally omitted, with a reason

Never silently turn an unknown value into a convenient value.

`missing` is not automatically `zero`.


# Quality has consequences

Poor data can cause:

- Incorrect financial reports
- Customers contacted twice—or not at all
- Inventory shortages
- Biased models
- Missed fraud
- Regulatory risk
- Loss of trust in all analytics

> **Discussion (4 min):** Which is more dangerous: obviously wrong data or believable-but-wrong data? Why?


# Why historical data matters

Current state answers:

> What is true now?

History answers:

- What changed?
- When did it change?
- How quickly is it changing?
- Is today unusual?
- What happened before an outcome?
- What might happen next?

Without history, there is no trend, comparison, seasonality, or learning.


# Overwrite or preserve?

Customer `C-208` moves from Chennai to Bengaluru.

| Approach | Stored result | Can answer “Where did the customer live when ordering?” |
|---|---|---|
| Overwrite | Bengaluru only | No |
| Preserve history | Chennai until move date; Bengaluru after | Yes |

Operational systems may only need the latest address.

Analytics often needs the value **as it was at the time of the event**.


# Two kinds of time

An event may have more than one important timestamp:

- **Event time:** when it happened in the real world
- **Processing time:** when the organization received or handled it

Example:

> A store completes a sale Monday at 23:58, but its connection recovers Tuesday at 00:12.

Does the sale belong to Monday or Tuesday?

The answer depends on the question—so both times may need preservation.


# History enables comparison

```text
Single snapshot:     Revenue = ₹4.2M
                     Is that good?

With history:        ₹3.1M → ₹3.4M → ₹3.8M → ₹4.2M
                     Growth is visible.

With context:        Last year same month = ₹4.5M
                     The interpretation changes.
```

A number without a baseline is often just a number.


# Why analytics stores data differently

Operational design optimizes for:

- Quickly finding and changing one record
- Avoiding conflicting updates
- Supporting a specific application

Analytical design optimizes for:

- Reading millions of records
- Grouping and summarizing
- Comparing long periods
- Combining business domains
- Making common questions simple

Different workloads lead to different organization and storage choices.


# From application model to business model

An application may separate data to update it safely:

```text
Orders → Order items → Products → Categories
```

Analytics may organize the same facts around a measurable event:

```text
                  Date
                   │
Customer ──── SALES EVENT ──── Product
                   │
                 Store
```

The center answers **what happened**; surrounding context explains **who, what, where, and when**.


# Facts, dimensions, and measures

For a sale:

- **Fact:** the sale occurred
- **Measures:** quantity, revenue, discount, cost
- **Dimensions:** date, customer, product, store, channel

Analytical questions combine them:

> **Revenue** *(measure)* by **month and region** *(dimensions)*.

A clear analytical model gives people a shared vocabulary for asking questions.


# Granularity: what does one row mean?

Possible meanings of one row:

- One order
- One item within an order
- One payment attempt
- One daily product total
- One customer-month summary

This meaning is the **grain**.

Mixing grains creates incorrect totals. Before using a dataset, ask:

> **What exactly does one row represent?**


# Detail versus summary

| Detailed data | Summarized data |
|---|---|
| Flexible for new questions | Fast and simple for known questions |
| Large volume | Smaller volume |
| More context | Some detail is intentionally lost |
| Requires careful use | Easier for broad consumption |

Organizations often keep both:

- Detail for investigation and new analysis
- Summaries for repeated reporting and rapid access


# Who consumes the data?

| Consumer | Typical need |
|---|---|
| Executives | Strategic indicators and trends |
| Managers | Team performance and exceptions |
| Analysts | Flexible exploration and explanation |
| Data scientists | Historical patterns and model inputs |
| Operational teams | Timely alerts and prioritized actions |
| Customers | Recommendations, status, personal insights |
| Regulators/auditors | Traceable, reproducible evidence |

Different consumers need different detail, freshness, language, and access.


# One dataset, different views

A delivery dataset can support:

- **Executive:** On-time delivery rate by quarter
- **Operations manager:** Late shipments by warehouse today
- **Analyst:** Delay distribution by carrier and route
- **Customer:** Where is my package?
- **Data scientist:** Features for predicting late delivery
- **Auditor:** Evidence for a disputed shipment

Good engineering starts with consumer needs while preserving reusable foundations.


# Trust requires more than correctness

Consumers also need to know:

- What does this metric mean?
- Who owns it?
- How fresh is it?
- Where did it come from?
- What changed recently?
- Who is permitted to see it?
- Can the result be reproduced?

This is the role of **documentation, lineage, governance, security, and observability** across the lifecycle.


# Responsibilities across the lifecycle

Data engineering is collaborative:

- **Source owners** explain how data is created
- **Domain experts** define business meaning
- **Data engineers** make movement and preparation reliable
- **Analysts/scientists** interpret and model
- **Governance/security teams** protect appropriate use
- **Consumers** validate whether outputs help decisions

Data quality cannot be “owned” by one team alone.


# Design backwards from the decision

Start with:

1. What decision will be made?
2. Who makes it?
3. What question supports that decision?
4. What definition and level of detail are needed?
5. How fresh and accurate must the answer be?
6. Which source events can produce it?

Then design the lifecycle.

> A pipeline without a consumer need is movement, not value.

# Scenario: a late delivery dashboard

A manager asks:

> “Show late-delivery rate by warehouse every morning.”

Before building anything, clarify:

- What counts as “late”?
- Promised time or customer-selected time?
- Which timezone?
- Are cancelled orders included?
- How are partial deliveries handled?
- What if carrier events arrive a day late?
- Can historical definitions change?

The hardest part is often agreeing on meaning—not moving bytes.


# Common misconceptions

❌ “Data from a system must be correct.”  
✅ Systems faithfully record both normal behavior and failures.

❌ “Real time is always better.”  
✅ Freshness should match the decision.

❌ “Cleaning means deleting bad rows.”  
✅ Quality issues need explicit, traceable treatment.

❌ “Current state is enough.”  
✅ Analysis depends on change over time.

❌ “One table can serve everyone.”  
✅ Consumers need different shapes, access, and detail.


# The lifecycle as a set of promises

```text
SOURCE      “This event means what we say it means.”
INGESTION   “We move it completely and predictably.”
STORAGE     “We preserve it safely.”
TRANSFORM   “We apply transparent, consistent rules.”
SERVING     “We make it usable for the intended audience.”
CONSUMER    “We interpret it within its limits.”
```

Trust emerges when every promise is visible and testable.

# Key takeaways

- Data is a contextual representation of reality
- Operational data runs the present; analytical data explains patterns
- Data comes from many partial, imperfect sources
- The lifecycle moves from generation to trusted consumption
- Cleaning makes quality explicit and use-appropriate
- History makes change, comparison, and learning possible
- Analytics stores data around large-scale reading and business questions
- Value is realized only when a consumer can make a better decision
