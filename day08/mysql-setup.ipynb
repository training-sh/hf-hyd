```python
import mysql.connector

ROOT_CONFIG = {
    "host": "localhost",
    "user": "root",
    "password": "root"
}

PASSWORD = "Te$M@987"

students = [
    "student01",
    "student01"
]

conn = mysql.connector.connect(**ROOT_CONFIG)
cursor = conn.cursor()

for database, username in students:

    cursor.execute(f"CREATE DATABASE IF NOT EXISTS `{database}`")

    cursor.execute(
        f"CREATE USER IF NOT EXISTS '{username}'@'%' "
        f"IDENTIFIED BY '{PASSWORD}'"
    )

    cursor.execute(
        f"GRANT ALL PRIVILEGES "
        f"ON `{database}`.* "
        f"TO '{username}'@'%'"
    )

cursor.execute("FLUSH PRIVILEGES")

conn.commit()

cursor.close()
conn.close()

print("Completed.")
```
