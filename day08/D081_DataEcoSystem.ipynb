# Our running example: **ShopNow**

ShopNow is an online retailer with:

- A website and mobile application
- Customers, products, orders, and payments
- Warehouses and delivery partners
- Marketing and customer-support teams
- Managers who want daily business reports
- Analysts who investigate trends

We will follow ShopNow’s data from creation to consumption.


# What is a data ecosystem?

A **data ecosystem** is the connected environment in which data is:

```text
created → collected → moved → stored → processed → shared → consumed
```

It includes:

- Data and its meaning
- People and organizational responsibilities
- Processes and quality rules
- Platforms and technologies
- Security and governance

An ecosystem is more than a collection of tools.


# The ecosystem at a glance

```text
┌──────────────┐    ┌─────────────────────────┐    ┌──────────────┐
│ DATA SOURCES │───►│     DATA PLATFORMS      │───►│  CONSUMERS   │
│              │    │                         │    │              │
│ apps         │    │ ingest → store → process│    │ people       │
│ databases    │    │       → organize        │    │ applications │
│ files        │    │                         │    │ models       │
│ devices      │    │ quality + governance    │    │ regulators   │
└──────────────┘    └─────────────────────────┘    └──────────────┘
```

Each section exists to serve a different responsibility.


# Three questions for any architecture

1. **Where does the data come from?**
2. **What must happen before it is useful?**
3. **Who needs it, and for what decision?**

These questions reveal:

- Required sources
- Suitable movement and processing patterns
- Storage and quality needs
- Necessary freshness and detail
- Appropriate outputs

> Start with meaning and purpose before selecting technology.


# Part 1 — Data Sources

<div style="padding:1.5rem;border-left:8px solid #0284c7;background:#e0f2fe;">
<h2 style="margin-top:0;">Where organizational data begins</h2>
<p style="font-size:1.2rem;">A source is any system, person, device, or organization that produces data we want to use.</p>
</div>


# Common categories of sources

| Category | Examples |
|---|---|
| Business applications | E-commerce, banking, HR, CRM |
| Databases | Orders, accounts, inventory |
| Files | CSV, spreadsheets, JSON, XML |
| Machine events | Logs, clicks, application events |
| Devices | Sensors, GPS, industrial equipment |
| External providers | Partners, market feeds, public data |
| Human content | Forms, surveys, documents, emails |

One organization commonly uses all of these.


# Business applications

Applications generate data while helping users complete work:

- A shopping application records orders
- A bank application records transfers
- A hospital system records appointments
- A university portal records enrollment
- A support system records customer tickets

Example order event:

```text
Order O-1057 | Customer C-42 | Product P-18
Quantity 2   | Amount ₹1,998 | Time 10:32
```

The application’s first purpose is to run the process—not to support every future analysis.


# Operational databases as sources

Applications commonly keep their current state in databases:

| Table | Example content |
|---|---|
| Customers | Name, address, contact details |
| Products | Description, price, category |
| Orders | Customer, time, status, total |
| Inventory | Product, location, available quantity |

These databases are optimized to support application actions safely and quickly.

Analytical systems usually **copy** needed data rather than burdening the live application.


# Files as sources

Files remain common because they are simple to create and exchange:

- A supplier sends a daily product CSV
- Finance uploads a monthly spreadsheet
- An application exports JSON events
- A government publishes a public dataset

Typical file questions:

- What format and character encoding?
- Does every file follow the same structure?
- Is it a full snapshot or only new records?
- How do we know a file is complete?
- Can the same file arrive twice?


# Logs and event data

Systems continuously record what happened:

```text
10:30:02  product_viewed   customer=C42  product=P18
10:31:14  added_to_cart    customer=C42  product=P18
10:32:08  order_created    customer=C42  order=O1057
10:32:10  payment_success  order=O1057
```

Events help reconstruct behavior over time.

They may be high-volume, repeated, late, missing, or out of order—so “generated automatically” does not mean “ready for analysis.”


# Devices and sensors

Connected devices produce measurements:

- Temperature every 10 seconds
- Vehicle position every minute
- Machine vibration continuously
- Smart-meter usage every 15 minutes

Example:

```text
sensor_id = WH-07
temperature = 8.4
unit = °C
measured_at = 10:35:00
```

Without the unit, time, device identity, and location, `8.4` has little meaning.


# External and partner sources

Organizations also receive data they do not control:

- Delivery status from logistics partners
- Payment results from payment processors
- Exchange rates
- Weather observations
- Demographic or geographic data
- Open government datasets

Important considerations:

**ownership · permission · reliability · update frequency · contract · meaning**


# Structured, semi-structured, unstructured

| Type | Typical form | Example |
|---|---|---|
| **Structured** | Fixed rows and fields | Orders table |
| **Semi-structured** | Labels with flexible structure | JSON event, log |
| **Unstructured** | Free-form content | Email, image, PDF, audio |

```text
Order row     → fields are predetermined
JSON event    → fields can vary by event
Support email → meaning is embedded in language
```

Modern ecosystems must manage multiple forms—not only tables.


# Source data changes

A source is not permanently stable:

- A new field is introduced
- A field becomes optional
- A status is renamed
- Date format changes
- An application is replaced
- A business definition evolves

This is called **schema or semantic change**.

> A pipeline can keep running technically while producing incorrect meaning.


# Source-system reality

Data may be:

- Missing
- Duplicated
- Delayed
- Incorrectly formatted
- Inconsistent across systems
- Recorded at different levels of detail
- Created using different definitions

Example:

> Sales calls a transaction “complete” after payment.  
> Finance calls it “complete” only after the return period closes.

Meaning must be resolved, not guessed.


# Activity — identify the sources

A food-delivery company wants to analyze late deliveries.

What sources might it need?

- Customer order application
- Restaurant preparation system
- Driver location events
- Map and traffic data
- Payment system
- Customer-support tickets
- Weather data

> **Discuss for 3 minutes:** Which source gives a fact, and which source only provides supporting context?


# Moving away from a source

Data can be collected in two broad ways:

| Batch movement | Continuous movement |
|---|---|
| Periodic groups of records | Events shortly after creation |
| Daily sales file | Live order-status event |
| Suitable when delay is acceptable | Suitable when rapid action matters |
| Often simpler | Often requires more coordination |

The right choice depends on **how quickly the consumer must act**.


# Source → movement → destination

```text
Order database ── periodic copy ──► analytical storage

Website events ── continuous flow ─► event platform

Supplier files ── scheduled arrival ► file landing area

Sensor readings ─ continuous flow ──► monitoring platform
```

The movement method can differ for each source, even inside one ecosystem.


# A small ETL/ELT orientation

```text
ETL: Source → Extract → Transform → Load → Target

ELT: Source → Extract → Load → Transform inside target
```

- **Extract:** obtain data from the source
- **Transform:** clean, standardize, join, calculate
- **Load:** place it in a destination

Both patterns turn source data into usable data. We focus today on the ecosystem around them, not their implementation.


# Part 2 — Data Platforms

<div style="padding:1.5rem;border-left:8px solid #7c3aed;background:#ede9fe;">
<h2 style="margin-top:0;">Where data is moved, stored, processed, governed, and served</h2>
<p style="font-size:1.2rem;">A platform provides shared capabilities so every team does not rebuild the entire lifecycle.</p>
</div>


# What does a data platform provide?

A modern data platform commonly supports:

- Connecting to sources
- Moving data reliably
- Storing raw and prepared data
- Batch and streaming processing
- Data-quality checks
- Cataloging and documentation
- Security and access control
- Serving data to consumers
- Monitoring failures and freshness

It may contain several technologies working as one system.


# OLTP — running the business

**Online Transaction Processing (OLTP)** supports many small operational actions.

Examples:

- Place one order
- Transfer money
- Reserve one seat
- Update one customer address

Typical characteristics:

- Fast reads and writes for a few records
- Many simultaneous users
- Current state
- Strong transactional correctness
- Data organized around application processes


# OLAP — understanding the business

**Online Analytical Processing (OLAP)** supports analysis across many records.

Examples:

- Revenue by quarter, region, and category
- Average delivery time over two years
- Customer-retention trends
- Product demand by season

Typical characteristics:

- Large scans and aggregations
- Mostly reading rather than frequent small updates
- Historical and integrated data
- Data organized around analytical questions


# OLTP versus OLAP

| | OLTP | OLAP |
|---|---|---|
| Purpose | Execute business processes | Analyze the business |
| Typical request | Find/update one order | Summarize millions of orders |
| Data | Current, application-specific | Historical, integrated |
| Users | Applications and operators | Analysts and decision-makers |
| Changes | Frequent small writes | Periodic/continuous prepared loads |

```text
Customer places order → OLTP
Manager studies annual sales → OLAP
```


# Why not analyze the operational system directly?

Heavy analytical queries can:

- Slow customer-facing transactions
- Compete for resources
- Require data that the application does not contain
- Need years of history the application overwrites
- Interpret fields differently across applications

```text
Operational systems ── copy + prepare ──► Analytical platform
      run today                               study time
```

Separation protects operations and makes cross-system analysis possible.


# Data warehouse

A **data warehouse** stores integrated, cleaned, structured, historical data for analytics.

```text
Orders ─┐
CRM ────┼─► clean + integrate ─► DATA WAREHOUSE ─► reports
Finance ┘
```

Strengths:

- Consistent business definitions
- Strong support for structured analysis
- Predictable reporting
- Historical comparison
- Easy aggregation by business dimensions

Think: **trusted business data, organized for questions.**


# Warehouse example

A sales warehouse might contain:

| sale_date | product | category | region | quantity | revenue |
|---|---|---|---|---:|---:|
| 2026-07-24 | P-18 | Audio | South | 2 | 1998 |

This supports questions such as:

- Revenue by month and region
- Top categories by quarter
- Average order value
- Year-over-year growth

The structure makes repeated business questions easier to express.


# Data lake

A **data lake** stores large amounts of data in its original or lightly processed form.

```text
tables ──┐
JSON ────┤
logs ────┼─► DATA LAKE ─► multiple processing needs
images ──┤
files ───┘
```

Strengths:

- Accepts many data types
- Retains detailed raw data
- Supports exploration and new use cases
- Scales to large volumes

Think: **flexible storage for diverse data.**


# A lake needs organization

A lake without management can become difficult to use:

- Unknown files
- Duplicate datasets
- Missing ownership
- Unclear quality
- Conflicting definitions
- Excessive or unsafe access

Useful lakes need:

**catalogs · metadata · naming · quality · security · lifecycle policies**

> Flexibility without governance creates confusion, not insight.


# Data lakehouse

A **lakehouse** aims to combine:

- The flexibility and scale of a data lake
- The reliability and analytical usability associated with a warehouse

```text
diverse source data
         ↓
 lake-style storage
         +
table management, consistency, governance
         ↓
analytics, data science, reporting
```

It is an architectural approach—not simply a new name for every data lake.


# Warehouse, lake, and lakehouse

| | Warehouse | Lake | Lakehouse |
|---|---|---|---|
| Main emphasis | Trusted structured analytics | Flexible diverse storage | Flexible storage plus managed tables |
| Data form | Mostly curated tables | Raw and varied | Raw through curated |
| Typical consumers | BI and analysts | Engineers and data scientists | Multiple analytical workloads |
| Governance | Usually strong | Must be deliberately added | Designed as a central concern |

Real organizations may use more than one approach.


# Platform layers

```text
┌────────────────────────────────────────┐
│ SERVING: reports, queries, data products│
├────────────────────────────────────────┤
│ CURATED: trusted business concepts      │
├────────────────────────────────────────┤
│ CLEANED: validated, standardized data   │
├────────────────────────────────────────┤
│ RAW: preserved source data              │
├────────────────────────────────────────┤
│ INGESTION: batch and event movement     │
└────────────────────────────────────────┘
```

Cross-cutting: **security · metadata · quality · monitoring · governance**


# Batch processing

Batch processing handles a bounded collection of data:

```text
Yesterday’s orders
        ↓
clean + join + summarize
        ↓
daily sales report
```

Good fit when:

- Data naturally arrives periodically
- The result is needed hourly, daily, or weekly
- Large work can run as a scheduled job
- Simplicity matters more than immediate response


# Stream processing

Stream processing handles an ongoing flow of events:

```text
order events → validate → detect condition → update/alert
     ...the flow continues...
```

Good fit when:

- Rapid detection or response matters
- Events arrive continuously
- Results must update frequently

Examples: fraud alerts, live shipment tracking, equipment monitoring.

Streaming is not automatically better; it carries additional design complexity.


# Batch and streaming can work together

```text
                   ┌─► live alert
Order events ──────┤
                   └─► historical storage ─► daily report
```

The same event can support:

- A near-real-time operational response
- A later historical analysis

Modern platforms often combine both patterns rather than choosing one for everything.


# Hadoop — the historical shift

**Apache Hadoop** popularized distributed storage and processing across clusters of ordinary machines.

Core ideas:

- Split very large data across multiple machines
- Move processing close to stored data
- Continue despite some machine failures
- Process large batches in parallel

Hadoop helped establish “big data” platforms. Its ecosystem influenced many systems used today.


# Spark — general distributed processing

**Apache Spark** is a distributed processing engine.

Conceptually:

```text
large dataset
     ↓ divide work
worker 1 | worker 2 | worker 3
     ↓ combine results
processed dataset
```

Common uses:

- Batch transformation
- Large-scale SQL analytics
- Data preparation
- Some streaming and machine-learning workloads

Spark is a processor, not the entire data ecosystem.


# Flink — continuous event processing

**Apache Flink** is a distributed processing engine strongly oriented toward stateful streams.

Useful concepts:

- Events are processed continuously
- Event time can differ from arrival time
- Processing may maintain running state
- Results update as new events arrive

Example:

> Maintain each customer’s spending during the last 10 minutes and flag unusual activity.

Flink can also process bounded data, but streaming is central to its design.


# Apache Beam — describing processing

**Apache Beam** provides a programming model for describing data-processing pipelines.

```text
read → transform → group → calculate → write
                 pipeline description
                          ↓
                  chosen execution engine
```

Important distinction:

- Beam expresses the processing logic
- A compatible **runner** executes it

Beam encourages similar concepts for bounded and unbounded data.


# Hadoop, Spark, Flink, Beam: roles

| Technology | Simple mental model |
|---|---|
| Hadoop | Distributed storage/processing ecosystem that shaped big data |
| Spark | General distributed engine, widely used for large-scale batch processing |
| Flink | Distributed engine centered on continuous, stateful stream processing |
| Beam | Portable model for describing batch and streaming pipelines |

These are not direct one-for-one replacements in every situation.

> Ask what role a technology performs before comparing product names.


# One conceptual platform architecture

```text
SOURCES                 PLATFORM                         CONSUMERS

Orders DB ──batch──┐    ┌──────────────┐
Files ─────batch───┼───►│ raw storage  │──┐
                   │    └──────────────┘  │   ┌───────────┐
Clicks ───stream───┼───► event flow ─────┼──►│ processing│
Sensors ──stream───┘                      │   └─────┬─────┘
                                         │         ↓
                                         └──► curated data
                                                    ↓
                              dashboards | analysis | alerts | models
```

The arrows represent responsibilities, not one mandatory toolset.


# Platform concerns beyond processing

A useful platform must also answer:

- **Quality:** Is the data complete and valid?
- **Metadata:** What does it mean?
- **Lineage:** Where did it come from?
- **Security:** Who may access it?
- **Privacy:** Is personal data handled appropriately?
- **Freshness:** When was it last updated?
- **Reliability:** What happens when part of the system fails?
- **Cost:** Is the value worth the resources used?


# Data catalog and metadata

A catalog helps people discover and understand data:

```text
Dataset: daily_sales
Meaning: completed sales by business day
Owner: Sales Analytics
Freshness: updated by 08:00 daily
Grain: one product, store, and day per row
Source: order and refund systems
Quality: certified
```

Finding a dataset is useful only when its meaning and limitations are clear.


# Part 3 — Data Consumers

<div style="padding:1.5rem;border-left:8px solid #16a34a;background:#dcfce7;">
<h2 style="margin-top:0;">The people and systems that turn data into action</h2>
<p style="font-size:1.2rem;">A platform has value only when consumers can use its outputs appropriately.</p>
</div>


# Who consumes data?

| Consumer | Main purpose |
|---|---|
| Executives | Strategy and organizational performance |
| Managers | Operational planning and exceptions |
| Analysts | Explore, explain, and communicate |
| Data scientists | Train and evaluate models |
| Applications | Personalize or automate experiences |
| Operational teams | Respond to alerts and current conditions |
| Customers | Track status and view personal insights |
| Auditors/regulators | Verify compliance and evidence |


# Business intelligence consumers

Dashboards and reports present repeatable measures:

- Daily revenue
- Monthly active customers
- Inventory availability
- Delivery performance
- Budget versus actual cost

Example:

```text
On-time delivery: 92.4%  ↓ 1.8 points
Worst region:      West
Main cause:        Warehouse processing delay
```

A useful dashboard connects an indicator to a possible action.


# Analysts

Analysts ask questions that evolve:

1. Did sales decline?
2. In which region?
3. Which category caused the change?
4. Was traffic lower or conversion weaker?
5. Did a promotion end?

They need:

- Detailed and summarized data
- Clear definitions
- Flexible exploration
- Historical context
- Ability to trace surprising results


# Data scientists and machine learning

Data scientists use historical data to discover patterns and build models.

Example delivery prediction inputs:

- Warehouse workload
- Destination distance
- Carrier
- Day of week
- Weather conditions
- Previous delay patterns

The model’s quality depends on representative, correctly timed, well-understood data—not only the algorithm.


# Applications as consumers

Applications can consume prepared data directly:

- Product recommendations
- Estimated arrival times
- Personalized home pages
- Search ranking
- Automated fraud checks

```text
historical behavior → prepared features → model/service → application response
```

Analytical data can return to the operational world and influence the next customer experience.


# Operational teams and alerts

Some consumers need action, not a dashboard:

```text
temperature event
       ↓
continuous processing
       ↓
threshold and context check
       ↓
warehouse alert
       ↓
technician investigates
```

A good alert is timely, understandable, and actionable. Too many low-quality alerts cause people to ignore them.


# Consumers need different views

The same delivery data can become:

- **Executive:** Quarterly on-time delivery trend
- **Manager:** Today’s delayed shipments by warehouse
- **Analyst:** Detailed events for root-cause analysis
- **Customer:** Current status of one package
- **Model:** Historical features for delay prediction
- **Auditor:** Traceable evidence for a disputed delivery

One underlying ecosystem can serve many products—but not necessarily through one identical table.


# Access should match responsibility

Not every consumer should see every field.

| Consumer | Appropriate access example |
|---|---|
| Executive | Aggregated regional results |
| Support agent | Details for assigned customers |
| Analyst | De-identified customer behavior |
| Finance | Required transaction details |
| Public user | Published aggregate statistics |

Security includes authentication, authorization, masking, auditing, and responsible use.


# Consumer needs shape the platform

| Need | Architectural implication |
|---|---|
| Monthly report | Scheduled batch may be sufficient |
| Fraud response in seconds | Continuous event processing |
| Flexible investigation | Detailed, queryable history |
| Regulatory evidence | Traceability and reproducibility |
| Customer-facing status | Low-latency serving |
| Model training | Historical, representative datasets |

Technology choices should follow these needs.


# End-to-end example: sales analytics

```text
Website orders ──┐
Store sales ─────┼─► ingest ─► raw storage
Product file ────┘                ↓
                       validate + standardize
                                  ↓
                         join + calculate
                                  ↓
                        curated sales data
                         ↙       ↓       ↘
                    dashboard  analyst  forecast
```

Each arrow can introduce delay, quality issues, and changes in meaning.


# End-to-end example: equipment monitoring

```text
Machine sensors ─► event flow ─► continuous processing ─► alert
       │                                  │
       └──────────────► history ◄─────────┘
                              ↓
                       reliability analysis
                              ↓
                    preventive-maintenance plan
```

The same source supports both immediate response and long-term improvement.


# Reading an architecture diagram

For every box and arrow, ask:

1. What data exists here?
2. Is it current state, events, raw history, or curated information?
3. Is movement batch or continuous?
4. What transformation occurs?
5. What happens if it fails?
6. Who owns the result?
7. Which consumer requirement does it serve?

Architecture diagrams are compressed explanations—not decoration.

# Common misconceptions

❌ “A platform is one database.”  
✅ It is a set of connected capabilities.

❌ “All data should stream in real time.”  
✅ Freshness should match the decision.

❌ “A lake automatically solves data problems.”  
✅ Flexible storage still needs governance.

❌ “Spark, Flink, and Hadoop are the same.”  
✅ They overlap but emphasize different roles.

❌ “Consumers only means dashboard users.”  
✅ Applications, models, teams, and customers also consume data.


# Choosing conceptually

Ask these before discussing technology:

- What decision or action is supported?
- Which data sources are required?
- What volume and variety exist?
- How fresh must the result be?
- Is the work bounded or continuous?
- Must raw history be preserved?
- What quality level is necessary?
- Who may access the data?
- How will failure be detected and recovered?

There is no universally “best” platform—only a suitable design for a defined need.

# The complete mental model

```text
SOURCES                  PLATFORM                       CONSUMERS
────────────────────────────────────────────────────────────────────
applications ─┐          ingest                         executives
databases ────┤          ↓                              managers
files ────────┼────────► raw → clean → curated ───────► analysts
events ───────┤          ↑       processing             models
devices ──────┤          batch + streaming              applications
partners ─────┘          quality + governance           customers
```

**Sources create evidence. Platforms make it trustworthy and usable. Consumers turn it into action.**


# Key takeaways

- A modern data ecosystem connects data, people, processes, and platforms
- Sources vary in form, ownership, speed, and reliability
- OLTP runs individual business transactions; OLAP analyzes patterns
- Warehouses, lakes, and lakehouses emphasize different storage needs
- Batch and streaming solve different timing requirements
- Hadoop, Spark, Flink, and Beam perform distinct ecosystem roles
- Consumers include humans, applications, models, and regulators
- Quality, metadata, governance, and security operate across every layer
- Architecture should begin with a decision—not a product name
