# Exercise — your assignment

You are a team of architecture interns at **TrustBank**, a new digital bank.

Design a **high-level data architecture** showing:

- How customers use the bank
- Where operational transactions are recorded
- How suspicious activity is detected
- How historical data reaches analytics
- How reports and other consumers receive data

Create one PowerPoint slide, export it as an image, and upload it.


# What “high-level” means

Show the major responsibilities and connections:

```text
users → channels → operational systems → data movement
      → analytical platform → consumers
```

Do show:

- Important systems and data stores
- Major flows and architectural boundaries
- Batch or streaming movement
- Consumers and responsibilities

Do not show:

- Individual database tables
- SQL statements
- Network ports
- Detailed deployment configuration


# Expected submission

Submit one readable architecture image:

```text
TeamName_TrustBank_Architecture.png
```

The image must contain:

- Architecture title
- Major components
- Labeled arrows
- Operational and analytical boundaries
- A small legend
- Two architectural decisions or notes

Keep the editable PowerPoint file in case your team is asked to revise the design.


# Team roles

Assign roles quickly:

- **Facilitator:** keeps the design moving
- **Business analyst:** checks requirements and consumers
- **Operational architect:** focuses on transactions and correctness
- **Data architect:** focuses on movement and analytics
- **Presenter/reviewer:** checks clarity and explains the result

For smaller teams, one person may take multiple roles.


# Scenario: TrustBank

TrustBank provides:

- Mobile and web banking
- Current and savings accounts
- Money transfers
- Debit-card payments
- Customer-support service
- Fraud monitoring
- Monthly statements
- Management and regulatory reports

The bank operates continuously. Incorrect balances or missing transactions are unacceptable.


# Business requirements

The architecture must support:

1. Customers viewing current balances
2. Customers transferring money
3. Card transactions arriving continuously
4. Fraud alerts within seconds
5. Monthly account statements
6. Daily management dashboards
7. Historical customer and transaction analysis
8. Regulatory reporting with traceable data


# Important data

| Data area | Examples |
|---|---|
| Customer | customer ID, contact, verification status |
| Account | account ID, type, status, balance |
| Transaction | transaction ID, amount, time, accounts |
| Card event | merchant, amount, location, result |
| Fraud alert | rule/model result, reason, review status |
| Support case | customer, issue, resolution |

You may add other data only when it supports a stated requirement.


# Consumers

Your design must serve:

- Customer
- Customer-support agent
- Fraud-monitoring team
- Operations manager
- Business analyst
- Finance team
- Regulatory reporting team

Optional:

- Fraud-detection model
- Customer-notification service

Ask what freshness, detail, and access each consumer needs.


# Architecture constraints

- A transfer must update related balances safely
- Customer-facing requests must remain responsive
- Analytical reports must not slow transaction processing
- Fraud detection needs near-real-time events
- Historical transactions must be retained
- Sensitive customer information must be restricted
- Data used for regulatory reporting must be traceable
- Failed data movement must recover without duplicating transactions


# Required architecture zones

```text
┌─────────────┐  ┌────────────────┐  ┌─────────────────┐
│ CHANNELS    │  │ OPERATIONAL    │  │ DATA MOVEMENT   │
│ mobile/web  │─►│ systems & OLTP │─►│ batch/CDC/stream│
└─────────────┘  └────────────────┘  └────────┬────────┘
                                              ▼
                         ┌───────────────────────────────┐
                         │ ANALYTICS & SERVING           │
                         │ history → prepare → consumers │
                         └───────────────────────────────┘
```

Adapt these zones to your design.


# Decision 1 — operational ownership

Decide which operational responsibility owns:

- Customer profile
- Accounts and balances
- Transfers
- Card-payment events
- Fraud cases
- Support cases

You may group related responsibilities, but avoid one unexplained box called “Bank Database.”

> A clear boundary states what a system owns and why.


# Transaction challenge

Customer A transfers ₹5,000 to Customer B.

The system must:

1. Verify available balance
2. Debit Account A
3. Credit Account B
4. Record the transaction
5. Produce confirmation

Add an architectural note:

> “Transfer processing uses a transaction to preserve balance correctness.”

No SQL or detailed lock design is required.


# Decision 2 — fraud path

Card and transfer events must reach fraud monitoring within seconds.

Show a path such as:

```text
transaction event
      ↓ continuous movement
fraud evaluation
      ↓
alert/case → fraud analyst
```

Decide whether the event also enters historical analytical storage—and show that path if it does.


# Decision 3 — analytical path

Daily dashboards and regulatory reports should not query the live transaction system heavily.

Show:

```text
operational data
      ↓ batch, incremental, or CDC
raw/landing area
      ↓ validate + integrate
analytical warehouse
      ↓
reports and analysts
```

Label the expected refresh frequency.


# Movement choices

| Flow | Likely approach |
|---|---|
| Fraud events | Continuous/streaming |
| Account and transaction changes | Incremental or CDC |
| Daily management reporting | Scheduled preparation |
| Monthly statements | Batch |
| Historical recovery | Full or bounded batch |

These are guidance—not mandatory answers.

Your arrows must identify both the data and movement style.


# Security and governance

Add at least three brief controls:

- Restrict access by job responsibility
- Encrypt sensitive data
- Mask customer details in general analytics
- Record who accessed regulated data
- Catalog important analytical datasets
- Preserve lineage for regulatory reports
- Define retention periods

Security should appear across the architecture, not only as one isolated box.


# Failure and recovery

Choose one failure and show or annotate the response:

- Duplicate transaction event
- Delayed analytical update
- Fraud processor temporarily unavailable
- Movement job fails halfway
- Analytical warehouse unavailable

Example note:

> “Events are retained and replayed; transaction IDs prevent duplicate results.”


# 60-minute plan

| Phase | Time |
|---|---:|
| Read scenario and assign roles | 5 min |
| Identify components and boundaries | 10 min |
| Draw operational and fraud paths | 10 min |
| Add analytical movement and consumers | 10 min |
| Add security, failure, and decision notes | 10 min |
| Review and export | 5 min |
| Team presentations/peer review | 10 min |

Move forward when a decision is reasonable; record assumptions instead of getting stuck.


# Design workflow

### 1. Draft

List components before using PowerPoint.

### 2. Connect

Draw the critical business and data flows.

### 3. Label

State what arrows carry and how quickly.

### 4. Challenge

Test transaction, fraud, and reporting scenarios.

### 5. Simplify

Remove details that do not explain an architectural decision.


# PowerPoint visual guide

Suggested colors:

| Color | Role |
|---|---|
| Blue | Channel or application |
| Green | Operational system/OLTP |
| Orange | Movement or processing |
| Purple | Analytical system/OLAP |
| Grey | Consumer |
| Red | Security, risk, or alert |

Use short labels, aligned boxes, directional arrows, and a legend.

One readable slide is the constraint.


# Architecture stress test

Trace these three journeys:

### Transfer

Customer → transaction processing → account update → confirmation

### Fraud alert

Card event → continuous evaluation → case → fraud analyst

### Regulatory report

Operational history → movement → validated warehouse → regulated output

If any journey stops without reaching its consumer, revise the diagram.


# Final checklist

- [ ] Mobile and web channels are shown
- [ ] Operational ownership is clear
- [ ] OLTP and analytical workloads are separated
- [ ] Transfer correctness is noted
- [ ] Fraud path supports rapid response
- [ ] Historical analytical storage is present
- [ ] Data-movement arrows are labeled
- [ ] All required consumers are connected
- [ ] Security and lineage are visible
- [ ] One failure-recovery approach is included
- [ ] The image is readable

# Three-minute team presentation

Explain:

1. Your major architecture zones
2. How a transfer remains correct
3. How fraud events reach analysts quickly
4. How operational data reaches historical analytics
5. One security or recovery decision

Peer reviewers ask:

- Which system owns this data?
- What does this arrow carry?
- What trade-off did the team accept?


# Reference pattern — reveal after submission

```text
Customers → Mobile/Web → Banking services → Accounts & Transactions OLTP
                                          │
Card network → card events ───────────────┤
                                          ├─stream→ fraud evaluation → alerts
                                          │
                                          └─CDC/incremental→ raw landing
                                                               ↓
Support system ───────── incremental ───────────────► validate + integrate
                                                               ↓
                                                    analytical warehouse
                                                     ↙      ↓       ↘
                                               dashboard analyst regulator
```

This is one defensible pattern, not the only correct solution.


# Reflection

After the presentations, record:

- One boundary your team explained well
- One arrow that needs a clearer label
- One security consideration you initially missed
- One difference between the fraud and reporting paths
- One improvement learned from another team

Architecture is refined through review, not completed in the first draft.


# Practice-session takeaway

```text
High-level data architecture =
clear responsibilities
  + correct operational boundaries
  + appropriate movement
  + separated analytical workloads
  + governed consumer access
  + explicit trade-offs
```

A good architecture lets technical and business audiences follow the same story.
